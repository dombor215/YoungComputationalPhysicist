# **Chapter 0: Welcome to SageMath!**




Physics has always been done with three instruments: your head, a pencil and a lot of paper. Galileo managed with those, Newton too, and Einstein. But somewhere between them and us, physics grew. Today's questions - the collision of two black holes, the weather over Central Europe tomorrow, the hundreds of millions of collisions inside the LHC each second, a quantum dot a few hundred atoms wide - have left the pencil behind. Not because the physics got harder to *think*, but because the *amount of it* got out of hand. The fastest human alive, with a calculator and infinite patience, could not finish the arithmetic that a modern physics question demands by lunchtime.

So today's physicist has a fourth instrument: the computer. Not as a replacement for the head, but as the head's drudge. Real physicists spend a striking part of their day *programming*: typing formulas for a machine to carry, because that is the only way to feed a real calculation - messy data, many coupled equations, a simulation run a thousand times - through to an answer.

And here is the catch for a beginner. To go near that world you seemed to need two trades at once: a physicist's and a programmer's. This book is about the fact that you don't. There is a tool that lets you do serious, modern computational physics while writing nothing but formulas - **SageMath** - and this chapter is a look at what it can do before we get down to work.



## **0.1 What is SageMath?**

SageMath is, honestly, hard to label. On the surface it is a *super-calculator* on which you type a formula and get an answer. One layer down it is *computer algebra software*: it keeps expressions symbolic and exact, in a similiar way you write them on paper. It draws publication-grade graphs, handles exact general solutions of equations, provides result with  arbitrary precission. And at its core it is the Python programming language, wearing physics clothes. When your work outgrows a single formula - and it will, the programming is already there under your fingers, waiting.

To go throught this book, you do not need to install anything, own a powerful computer, or know any programming to follow this book. You mostly need a browser, some sense to observe patterns and eagerness to learn some new skills that are applicable beyond physics.

> **Note: where to run SageMath.** For simple, fast computations the quickest door is a SageMath cell: open https://sagecell.sagemath.org/ in a browser, type, evaluate. Our SageCell IDE at https://sage-sk.org/IDE/ offers the same cells in a more advanced mode. When your problem solving grows into something continuous, a *computational essay* mixing text, code and results, move to Jupyter notebooks. If you prefer the cloud, where the computations run on a web server, there is CoCalc (paid) and our SageMath AI Notebooks (free, suitable for lightweight work, you are already look at one right know, for more details and updates see https://sage-sk.org/). To install SageMath on your own computer, follow the instructions at https://www.sagemath.org/; on Windows the installation of SageMath (and full Jupyter environment alongside it) needs a little tweaking with WSL (Windows Subsystem for Linux), see https://doc.sagemath.org/html/en/installation/index.html. You can give it a try once you will deal with something more serious in the future on your own.

## **0.2 The body of the notebook - cells**

Every notebook is built from a series of **cells** - little chunks of content, either explanations (such as this one) or code.

* **Text cells** carry the accompanying words for your computations. They are written in the simple *Markdown* formatting language, and they also support $\LaTeX$, which typesets great-looking mathematical formulas, e.g.:
    * Euler's identity: $e^{i\pi} + 1 = 0$
    * the Schrödinger equation:
$$i\hbar \frac{\partial}{\partial t}\Psi(\vec{r},t)
=
\left[
-\frac{\hbar^2}{2m}\nabla^2 + V(\vec{r},t)
\right]\Psi(\vec{r},t)$$
    * one you will meet again in Chapter 2, the trajectory of a kicked ball: $y = x\tan\alpha - \dfrac{g\,x^2}{2v_0^2\cos^2\alpha}$

* **Code cells** contain executable Sage commands. The most important thing you can do with them is run them:



```sage
# a line starting with a hash sign is a comment, ignored by Sage
# store a greeting into the variable "greeting"
greeting = "Hi there, this is Sage!"
```

```sage
# display the value stored in the previous cell
print(greeting)
```


Run a cell with the Execute button, or with `Ctrl + Enter` (`Shift + Enter` runs it and moves on to the next cell). When a code cell runs, the commands are executed and either **the result is displayed** below the cell, or the data you entered (e.g. variables) **is stored in memory**.

> **Note: learning Markdown in the notebook.** Every text cell carries a small toolbar above it. Its **Edit** button reveals the raw Markdown source behind the rendered text, so you can see how any paragraph in this book was written, and the small **lightbulb** button opens a Markdown cheatsheet: a collection of ready-made snippets you can copy, paste into a cell and edit to your taste. The rendered result appears instantly below the cell, so you immediately see whether your Markdown did what you intended. One more tip for comfortable typing: set your keyboard to the **US English layout**. Programming, Markdown and $\LaTeX$ were all born on it, and symbols like `/`, `[`, `]`, `\` or `~` sit exactly where this book's code expects them, which makes writing code and Markdown considerably smoother.

The notebook remembers. Code cells need to be **run in sequence**, starting from the first, so that intermediate results from earlier cells are available in the later ones. If you change a cell, **re-run** it: the new values overwrite the old ones and the outputs update.

## **0.3 A first taste: Sage as a physicist's assistant**

> ***Note:** Do not worry if not everything in this section is clear yet. With a feel for patterns and formulas (plus a bit of English) you can work out what the individual commands mean. Try making small changes in the code and see how they play out. This chapter is a demonstration of what Sage can do for you; you will get to know everything properly in the chapters that follow.*

Sage's favorite trick is the one that separates it from any pocket calculator: it treats formulas the way paper does. Store Newton's second law - not the numbers, the *law*:



```sage
# typical things in Sage: example of Newton's 2nd law
var("F m a")                    # declare symbolic variables
newtons_second_law = a == F/m  # store the law as an equation (double equals sign)
```

```sage
# pretty display of the stored law
show(newtons_second_law)
```

```sage
# solve for the unknown m (mass)
show(solve(newtons_second_law, m))
```


One stored equation, any unknown - and it scales up without effort. A system of three linear equations with three unknowns $x, y, z$:

$$\begin{aligned}
2x + y - z = 3 \\
-x + 3y + 2z = 7 \\
4x - 2y + z = 1
\end{aligned}$$



```sage
# solve the system, four lines of code
var('x y z')
eq1 = 2*x + y - z == 3
eq2 = -x + 3*y + 2*z == 7
eq3 = 4*x - 2*y + z == 1
show(solve([eq1, eq2, eq3], x, y, z))
```


Or take physics back in time: the equation of motion for a body thrown upward from a height $h_0$ with speed $v_0$,

$$h = h_0 + v_0 t - \frac{1}{2}g t^2,$$

and let Sage find the general formula for the times at which the body reaches a given height $h$ - rearranging that takes half a page by hand:



```sage
# enter the equation of motion
var("h h0 v0 t g")
equation = h == h0 + v0*t - 1/2*g*t^2
show(equation)
```

```sage
# general solution for the time t
show(solve(equation, t))
```

```sage
# substitute numerical values into the second solution
values = [h0==100, v0==10, g==9.81, h==0]
show(solve(equation, t)[1])
show(solve(equation, t)[1].subs(values))
```


Sage also cleans up after you, on expressions no pocket calculator would touch. Simplify, for instance, the expression $\sqrt{\dfrac{m^2 v^4}{m^2 v^2} \cdot x}$:



```sage
# enter the expression
var("m v x")
expression = sqrt((m^2 * v^4)/(m^2 * v^2)*x)
show(expression)
```

```sage
# the expression simplified even further
show(expression.canonicalize_radical())
```


None of this is typing practice. Each of these moves - store the formula, solve for what you want, substitute numbers, simplify - is a move a working physicist makes daily, and each would cost minutes on paper, with every minute a chance for a slip. In Sage there is nothing to re-check: the steps that survive are exactly the ones that cannot go wrong. What remains your job - whether the physics you put in is the physics you meant - is the part worth your head.



## **0.4 Graphs as a first-class product**

Enter a function and Sage draws it, precisely and instantly:



```sage
# enter a function
f(x) = 1/x*sin(x)
show(f)
```

```sage
# a simple plot of f for x from -30 to 30
show(plot(f, (x,-30,30)))
```


> ***Notice:** Although we formally entered $f(x) = \dfrac{1}{x} \cdot \sin x$, Sage automatically simplified it to $f(x) = \dfrac{\sin x}{x}$. It can do this on a much larger scale with far more complicated expressions too.*

The same plot with "pro" formatting, ready to go into a lab report or even professional scientific article:



```sage
# the previous plot, formatted like a publication figure
plot_f_formatted = plot(f, (x,-30,30), color="magenta")
# colors are entered in English, e.g. "magenta", "red", "green", ...
show(plot_f_formatted,
     title=r"Plot of the function f = \dfrac{\sin x}{x}",
     frame=True, gridlines=True)
```


Where this stops being a calculator and becomes a physicist's tool is comparison: several variants of one phenomenon in one figure, with a legend. Here is damped oscillation, $x(t) = A\,e^{-bt}\cos(\omega t)$, under four different settings of amplitude, damping and frequency:





```sage
# definition of damped oscillations depending on time t
# and the parameters A, b, omega
x(t, A, b, omega) = A * e^(-b*t) * cos(omega*t)
show(x)
```

```sage
# curve 1: basic parameters (A=1, b=1, omega=1)
plot_1 = plot(x(t, 1, 1, 1), (t, 0, 10),
              legend_label=r"$A, b, \omega$ = 1, 1, 1")

# curve 2: increased frequency (omega=5)
plot_2 = plot(x(t, 1, 1, 5), (t, 0, 10),
              color="red",
              legend_label=r"$A, b, \omega$ = 1, 1, 5")

# curve 3: reduced damping (b=0.3)
plot_3 = plot(x(t, 1, 0.3, 1), (t, 0, 10),
              color="green",
              legend_label=r"$A, b, \omega$ = 1, 0.3, 1")

# curve 4: increased amplitude (A=2)
plot_4 = plot(x(t, 2, 1, 1), (t, 0, 10),
              color="cyan",
              legend_label=r"$A, b, \omega$ = 2, 1, 1",
              frame=True, gridlines=True,
              title="Damped oscillation for various parameters",
              axes_labels=[r"$t$", r"$x(t)$"])
```

```sage
# combine all the plots into one
plot_final = plot_1 + plot_2 + plot_3 + plot_4

# display the final plot
show(plot_final)
```

```sage
# save the plot in HD quality (300 DPI) as a PNG file
plot_final.save("damped_oscillations.png", dpi=300)
```


Change a parameter, re-run, and the figure redraws. That rhythm - *formula, plot, adjust, compare, save* - is the daily bread of physics from the teaching lab to the research paper, and in Sage it takes a few cells. (The chapters that follow work the same way: motion in Chapter 2, more plotting in Chapter 3, energy in Chapter 4.)




## **0.5 Why Sage, and why in this book?**

The honest competition first. Paper, pen and a calculator do fine for a quadratic equation. A spreadsheet (e. g. MS Excel) will draw a graph, if you first compute a table of its points by hand. GeoGebra draws graphs nicely, but each parameter change is another round of clicking.

Then the work grows. The data from a real experiment - Chapter 5 - arrives as a spreadsheet of a hundred measurements with uncertainties attached, and no calculator has a button for that. Loading it into a table, checking it for holes and typos, drawing it with error bars, fitting a model to it and reading the physics out of the fit - all of it Chapter 5 - are tasks where the pencil does not come back. This is the arithmetic that fills the days of modern physics, and it is exactly the arithmetic computers eat.

SageMath is our vehicle for that journey because it meets a beginner where they stand:

* **it reads like math.** `sqrt(2*h/g)` is the formula, not a program wrapped around it.
* **it is exact when you want exact** and numerical when you ask - and it knows the difference (Chapter 1 is devoted to this).
* **it is open and free**, running in any browser through the links above, or you can install it without any fees
* **it does not hide the road ahead.** Sage is built on Python, the language of scientific computing, so what starts as a super-calculator typing trivial code grows, without changing tools, into the skills real research runs on.

One honest warning before the first chapter: Sage will not think for you. It carries the arithmetic flawlessly, but what to compute, and whether the answer is sensible, stays your job. Chapter 1 starts exactly there: a stone, a well, and why the number of digits you report is a physics decision, not a calculator's.
