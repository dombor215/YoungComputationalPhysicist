# **Chapter 1: The Stone in the Well**



## **1.1 Free fall in one formula**

Drop a stone and it does not float: it falls faster and faster the longer it drops. Galileo worked out the rule more than 400 years ago, and it is remarkably simple. If we ignore air resistance, an object dropped from rest covers the height

$$h = \frac{1}{2}\, g\, t^2$$

after time $t$, where $g = 9.81\ \mathrm{m/s^2}$ is the gravitational acceleration at the Earth's surface (may change slightly with altitude and latitude). Turning the formula around, the time needed to fall through a height $h$ is

$$t = \sqrt{\frac{2h}{g}}.$$

Doubling the depth of a fall does not double the time: it grows like the square root: $t \propto \sqrt{h}$. A stone falling 90 m takes only about 1.4 times longer than one falling 45 m.

**Units convention.** In this whole book we work in SI units: meters, seconds, kilograms. We do not attach units to numbers inside Sage. Instead we write the unit into the variable name or a comment, and into axis labels on plots. So a depth is stored as `h = 45` with the understanding that this means 45 meters.



## **1.2 Meet Sage: exact and approximate numbers**

Before we solve the well problem, two minutes on the tool itself. SageMath here runs in a notebook of cells: you type into a cell, press Shift+Enter, and the result appears below. The same cells greet you wherever you run it, and there are several places to run it:

> **Note: where to run SageMath.** For simple, fast computations the quickest door is a SageMath cell: open https://sagecell.sagemath.org/ in a browser, type, evaluate. Our SageCell IDE at https://sage-sk.org/IDE/ offers the same cells in a more advanced mode. When your problem solving grows into something continuous, a *computational essay* mixing text, code and results, move to Jupyter notebooks. If you prefer the cloud, where the computations run on a web server, there is CoCalc (paid) and our SageMath AI Notebooks (free, suitable for lightweight work, see https://sage-sk.org/). To install SageMath on your own computer, follow the instructions at https://www.sagemath.org/; on Windows the installation needs a little tweaking with WSL (Windows Subsystem for Linux), see https://doc.sagemath.org/html/en/installation/index.html.

Whichever door you choose, the first cells are plain arithmetic, and it works the way paper does. Multiplication and division bind before addition and subtraction:



```sage
# precedence of * and / over + and -
2 - 1 + 3*4
```


Parentheses group exactly as on paper. Only round parentheses are used, and they must be closed:



```sage
(2 + 3)*4
```


Powers are typed with `^` (Shift+6 on a US keyboard) or with two asterisks, whichever your keyboard prefers:



```sage
2^10, 2**10
```


Both give `1024`. A small typing matter before we go on: Sage will also meet text in your cells - names of files, comments in quotes, later whole columns of data - and text is always wrapped in quote marks. Python, on which SageMath is built, accepts both the double `"` and the single `'` quote: `"hello"` and `'hello'` are the same string. The only rule is consistency: a string must start and end with the same mark, so `'hello"` is an error. This book mostly uses double quotes, single ones appear where they read better.

And now the thing a (relatively basic) pocket calculator cannot do properly:



```sage
sqrt(2)
```


Sage answers with `sqrt(2)`, not with 1.4142135... It keeps the value *exact*, as a symbol. That is useful, because exact things stay exact:



```sage
sqrt(2)^2
```


The answer is exactly `2`, not 2.0000000000000004. A normal calculator would have rounded along the way and lost that. If we want a decimal, we ask for one:



```sage
N(sqrt(2))
```

```sage
round(N(sqrt(2)), 4)
```


> **Syntax box: `N()` and `round()`.**
> `N(expr)` converts an exact expression into a decimal number. You can ask for a specific number of digits with `N(expr, digits=30)` (try `N(pi, digits=100)` for fun), and `.n()` after an expression does the same: `sqrt(2).n()`. `round(x, n)` rounds the decimal `x` to `n` digits after the point. Rule of thumb: stay exact as long as possible, go decimal at the end.

Fractions behave the same way. Sage does not turn $1/3$ into 0.3333...:



```sage
1/3 + 1/6
```


The answer is `1/2`, exactly. In ordinary Python this would come out as `0.5` only after rounding errors; in Sage exact arithmetic is the default. Of course we can also see it as a decimal:



```sage
N(1/3 + 1/6)
```


**Sage simplifies on its own.** It does not just store exact expressions, it tidies them up. The sum of fractions above came back already reduced to `1/2`, before you could reach for a pencil:



```sage
50/100
```

```sage
sqrt(8)
```


The fraction comes out as `1/2`, and the root as `2*sqrt(2)` - the largest perfect square is pulled out of the radicand. You would have done both simplifications by hand on paper; Sage does them without being asked.

One warning about decimal points, because it bites every newcomer once: Sage uses a **decimal point**, never a comma. `9,81` is not a number. And the moment a single decimal number enters a calculation, the whole result becomes approximate:



```sage
1/3 + 1/6      # exact
```

```sage
0.33 + 0.17    # approximate: a decimal point appeared
```


The first cell answers `1/2`, a fraction. The second answers a decimal number: from the moment the decimal point appeared, Sage stopped doing exact arithmetic and switched to machine numbers of finite precision. Two tricks steer this behavior when you need to:

> **Note: forcing exact or approximate.** If you want approximate results without typing `N()` everywhere, write at least one input value as a decimal: `10.0` instead of `10`. If you want the opposite, exact results from numbers that are naturally decimal, write them as fractions: `981/100` instead of `9.81`. Then only integers are involved, and Sage stays exact until you call for a decimal. The fraction trick must be applied to *all* decimal numbers entering the calculation; one decimal anywhere spoils the exactness.




### **1.2.1 Summary: arithmetic in Sage**

Before moving on to the well problem, let us briefly summarize everything you will need for calculations in Sage. You can also use this overview later as a cheat sheet.

**Basic operations.** The characters `+`, `-`, `*` and `/` are used. Sage treats the division `/` as a fraction and automatically reduces it to lowest terms:



```sage
2/64 + 5/32
```

```sage
18.4 - 3*7
```


The order of operations is the same as on paper. For grouping, **always use round parentheses** `(` `)`. Square `[ ]` and curly `{ }` brackets have a different meaning in Sage (lists, sets), so nested brackets from paper must be rewritten as round ones. Try to evaluate:

$$\dfrac{\left(1.5 + \frac{2}{3}\right) \cdot \left[4.75 - \left(2.25 + \frac{1}{2}\right)\right]}{5.6 + \frac{7.2}{3.4 - \frac{3}{4}}}$$



```sage
# space for the calculation (the correct result is approximately 0.521022383545069)

```


**Integer division and remainder.** Besides `/` there are two useful operators: `//` gives the integer part of the quotient and `%` the remainder after division:



```sage
17//5, 17 % 5
```


The result `(3, 2)` means $17 = 3\cdot 5 + 2$. You will also meet `factorial(5)` ($5! = 120$) and `abs(-3)` (absolute value).

**Exact and approximate numbers.** An approximate value is obtained with `N(expression)` or `(expression).n()`, and the number of significant digits is set with `digits=`. The functions `RR()` and `QQ()` convert a number to a real or a rational one, respectively:



```sage
N(22/7 + 1, digits=3), RR(22/7 + 1), QQ(-1.25)
```


The last value is displayed as the fraction `-5/4`. For irrational numbers (e.g. $\sqrt{2}$), `QQ()` of course cannot help, since they cannot be written as a fraction.

**Mathematical constants.** The number $\pi$ is written `pi` and Euler's number $e$ is written `e`. In Sage they behave **as symbols**: they stay in exact form until you ask for an approximate value. Sage can compute them to arbitrary precision:



```sage
(3.5*pi + 12*8)/5
```

```sage
N(pi, digits=100)
```


**Powers and roots.** A power is written `^` or `**`. Sage handles even huge numbers that a calculator could not cope with (try `2^10000`). The square root is `sqrt()`, and higher-order roots are written with a fractional exponent, $\sqrt[n]{x} = x^{1/n}$:



```sage
sqrt(256), 256^(1/4)
```


A root stays exact until you use `N()`. Try `sqrt(2)` and `N(sqrt(2))`. Beware: the exponent `1/4` must be in parentheses, otherwise `256^1/4` would mean $(256^1)/4$.

**Elementary functions.** An overview of the notation:

| Function | Sage notation |
|---|---|
| $\sin x$, $\cos x$, $\tan x$ | `sin(x)`, `cos(x)`, `tan(x)` |
| $\sin^{-1} x$, $\cos^{-1} x$, $\tan^{-1} x$ | `arcsin(x)`, `arccos(x)`, `arctan(x)` |
| $x^n$ | `x^n` or `x**n` |
| $\sqrt{x}$ | `sqrt(x)` |
| $\sqrt[n]{x}$ | `x^(1/n)` |
| $e^x$ | `exp(x)` or `e^x` |
| $\ln x$ | `log(x)` |
| $\log_a x$ | `log(x, a)` |
| $\log x$ (base 10) | `log(x, 10)` |

The argument of trigonometric functions is an angle **in radians**. Sage returns exact values, not rounded ones:



```sage
sin(pi/4)
```


The result `1/2*sqrt(2)` is exactly $\frac{\sqrt{2}}{2}$. You get an approximation with `N(sin(pi/4))`. If you need to enter an angle in degrees, convert it to radians: `sin(30*pi/180)` gives exactly `1/2`.

> **Summary.** Sage computes exactly until a decimal point appears in the calculation or until you ask for `N()`. Use the decimal **point**, round parentheses, and radians for angles.



## **1.3 The problem**

Here is a real situation. You are standing at the mouth of an old hand-dug well in the mountains. A sign (and a long rope) tells you the water surface is 45 m below you. You drop a stone in.

Two questions, both answerable before you let go:

1. How long do you have to wait until the stone hits the water?
2. How fast is the stone moving when it gets there? (Relevant if it were your phone instead of a stone.)

*Fig. 1.2: The mouth of a hand-dug well. A drop and a stopwatch are all the physics instruments needed to measure its depth.*

We will answer both with Sage, one small step at a time, checking every result before taking the next step. That is exactly how a working physicist uses a notebook: never type a long formula, run it, and hope.



## **1.4 Solution, step by step**

**Step 1: store the numbers we know.**

Physics first, precision later: start with a rounded value of $g$,



```sage
g = 10     # m/s^2, rounded
h = 45     # m
```


> **Syntax box: variables.**
> `g = 10` gives the name `g` to the number 10. From now on, typing `g` means 10. Names are your choice, but good names (`g`, `h`, `t`) make the notebook read like physics instead of like code. A line starting with `#` is a comment and is ignored by Sage, so we could also have written `g = 10  # m/s^2`.
> Note also that both numbers are *integers*. Watch what that buys us in the next steps.

**Step 2: the formula, typed in one small piece at a time.**

The formula says $t = \sqrt{2h/g}$. Rather than typing the whole thing at once, first build the inside:



```sage
2*h/g
```


Sage answers `9` - exactly, because only integers went in. A rough sanity check in the head: 90 divided by 10 is 9, yes.

> **Note: The units work out.**
> $2h/g$ has units of $\mathrm{m}/(\mathrm{m/s^2}) = \mathrm{s^2}$, so its square root is a time. Always check units on intermediate results; it catches most formula mistakes.

**Step 3: take the square root.**



```sage
t = sqrt(2*h/g)
```

```sage
t
```


The stone falls for exactly 3 seconds. Suspiciously clean - and that cleanliness is our own doing: it came from the rounded $g = 10$. Time for the realistic value:



```sage
g = 9.81           # acceleration in m/s^2
t = sqrt(2*h/g)    # impact time
t                  # displaying the result
```


**Step 4: read the number, but not all of it.**

`t` now shows something like `3.02891266407691`. Tempting to copy all of it; resist. Sage can print 15 digits (or any number you choose), but you cannot *measure* that precisely. The depth is known to two digits, so an answer of 3.02891266 s would be nonsense. How many digits deserve to be reported is part of physics too; we will return to it properly in the chapter on uncertainties. The stone falls for about 3.03 seconds. For a prettier look at a result there is also `show(t)`, which prints textbook typesetting - fractions as stacked bars, roots as radical signs, `pi` as $\pi$; try `show(1/3 + 1/6)` and compare with the raw output.

Did you notice what changed with the second `g`? The first was `10`, an integer, so everything stayed exact and `sqrt(9)` was `3`. The second is `9.81`, a number with a decimal point, so the whole result became approximate at once. One decimal anywhere in the calculation is enough.

**Step 5: round it for the answer.**



```sage
round(t, 2)
```


Answer to question 1: **about 3.03 s**. We round to three digits because the depth (45 m, two digits) and $g$ (three digits) do not justify more precision. Carrying ten digits around would be pretending we know more than we do.

**Step 6: the impact speed.**

The stone starts from rest and gains speed steadily: $v = g\,t$. We already have `t` in memory, so reuse it:



```sage
v = g*t
```

```sage
round(v, 1)
```


About 29.7 m per second. In units that mean more for a car driver:



```sage
v*3.6
```


Multiplying m/s by 3.6 converts to km/h: the stone hits the water at roughly **107 km/h**. That is answer to question 2, and a good reason not to drop your phone down the well.



## **1.5 The reverse problem: timing the fall**

Now flip the situation. You visit a second well with no sign and no rope, but you do have a stopwatch. You drop a stone and measure a fall time of 2.5 s. How deep is this well?

From $h = \frac{1}{2}gt^2$:



```sage
t_measured = 2.5
```

```sage
h_found = g*t_measured^2/2
```

```sage
h_found
```


The well is about 31 m deep. Note what happened here: we used Sage as a *reverse* calculator. In school problems you are usually given the depth and asked for the time. With a stopwatch, reality gives you the time, and the depth is the unknown. A physicist runs the formula in whichever direction the data allow.

One caution for real measurements: your stopwatch reading includes your reaction time, about 0.2 s, and the time the *sound* of the splash needs to climb back up (about 0.13 s for 45 m, since sound travels 340 m/s). We will meet both effects again in the exercises.

Typing the formula fresh every time we change a number is no way to work. Sage can store the formula itself, as a **symbolic function**:

$$t(h) = \sqrt{\dfrac{2h}{g}}$$

Read it literally as "the time $t$, depending on the height $h$":



```sage
t_h(h) = sqrt(2*h/g)   # a function of h only; g is remembered
show(t_h)
```


Now the depths answer on demand:



```sage
show(t_h(45))
```

```sage
show(t_h(10))
```


And results can be combined directly:



```sage
N(t_h(20) - t_h(10))
```


That last number is how much longer a 20 m drop takes than a 10 m drop. Notice the name we gave the function: `t_h`, not `t`. There is a reason, and it is a small lesson in notebook hygiene.

> **Note: names get overwritten.** Somewhere above, the number `3.02891266407691` was stored under the name `t`. Had we now written `t(h) = sqrt(2*h/g)`, that number would be gone, replaced by a function; earlier cells using `t` would silently mean something different after re-running. When a quantity changes *kind* (number to function), change its name too: `t_h(h)` for $t(h)$, `t_hg(h, g)` for $t(h, g)$. `time_in_minutes` would also work, but frequently used quantities deserve compact, readable names.

The function `t_h(h)` holds `g` by reference, not by value. If you later change `g` and re-run the function definition, the new `g` applies; if you change `g` and *don't* re-run the definition, the old value is still baked in. And the function can take more than one argument:



```sage
t_hg(h, g) = sqrt(2*h/g)   # t as a function of two variables
t_hg(45, 10).n()           # .n() at the end works like N()
```


With `g = 1.62` (the Moon) available as an argument, this one line answers Exercise 3 as well. Functions can also be modified wholesale:



```sage
t_minutes = t_h/60         # the same times, in minutes
show(t_minutes)
```

## **1.6 Many depths at once: lists**

So far each command produced one number. Physics rarely works in ones. Suppose you want the fall time for six depths at once. First put the depths in a **list**:



```sage
# the list of depths for which we want the times
depths = [5, 10, 15, 20, 25, 30]
results = [t_h(h).n() for h in depths]
results
```


A list is built with square brackets `[` `]`, its items separated by commas - one more reason for the decimal point, not the comma, in numbers. The line `[t_h(h).n() for h in depths]` reads, literally: *for each `h` in the list `depths`, compute `t_h(h).n()` and collect it into a new list named `results`.* The keywords `for` and `in` belong to the so-called **for loop**. This compact form is called a *list comprehension*, an elegant Python idiom. If you already know some Python, here is the same computation spelled out in the traditional way:



```sage
depths = [5, 10, 15, 20, 25, 30]    # the list of values to go through
results = []                        # a prepared empty list

for h in depths:                    # the traditional for loop
    value = t_h(h).n()
    results.append(value)           # .append() attaches a value to the list

results
```


One caution, the same lesson as with `t`: inside those cells, `h` is no longer the depth of our well. Defining `t_h(h)` turned it, and `g`, into *symbolic variables*. The list comprehension then used `h` as its running name. The number 45 survives inside the function only as a value you feed it. Nothing is lost, but names have drifted; this is why new quantities get new names.

Working with a list means pulling values back out of it, and that is done with **indices**:



```sage
results[0]      # the first value in the list
```

```sage
last_value = results[-1]   # the last value in the list
last_value
```


Indices count like this: the 1st value has index `0`, the 2nd has index `1`, the 3rd has index `2`, and so on. Python (and therefore SageMath) starts counting at `0`, which looks strange at first but pays off with negative indices: the last value has index `-1`, the second-to-last `-2`, so you never need to know the length of a list to reach its end. The 6th value of `results` sits at index `5` - and yes, in Python we would say "the sixth value has index five" without blinking.

Values from a list work in further calculations like any number:



```sage
results[2]/60     # the third value, converted to minutes
```


And lists grow without effort. Thirty calculations are no problem:



```sage
depths30 = [5, 6 .. 30]   # a range: 5 to 30 in steps of 1 (the step of the second value)
results30 = [t_h(h).n() for h in depths30]
results30[5:10]          # the 6th through 10th values; in a slice, the last index is excluded
```

```sage
results30[10]
```


Long lists like `[5, 6 .. 30]` are written with two dots between the first two values; the step is taken from the gap between them. A slice `results30[5:10]` is a list cut out of a list, and again the right-hand index is not included: it gives entries 5, 6, 7, 8, 9.



## **1.7 What you can do now**

- Use Sage as a calculator that keeps exact values until you ask for decimals with `N()`, and simplifies fractions and roots on its own.
- Store numbers under names (`g = 9.81`) and build formulas from them.
- Round sensibly with `round(x, n)` at the end, not in the middle.
- Solve a physics problem forwards (depth gives time) and backwards (time gives depth).
- Store a formula itself as a symbolic function (`t_h(h) = sqrt(2*h/g)`), call it with any values, and modify it (`t_h/60` for minutes).
- Turn one formula into a whole table of results with a list comprehension (`[t_h(h).n() for h in depths]`), and read values back out with indices, counting from `0`.

We promised no programming, and a list comprehension may look like programming - it is. But it earned its place the same way every other command here did: one physics question, many inputs. From here on, a formula and a list of measurements are all we need.
