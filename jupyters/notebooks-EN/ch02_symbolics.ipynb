## **2.2 The physics of the flying ball**

Before SageMath gets anything to chew on, here is the physics behind the kick in words, because every formula in this chapter grows from one old idea.

A kicked ball is a *projectile*: an object that got one push and afterwards flies on its own, with nothing steering it. Cannonballs, basketballs and the water of a fountain are projectiles too. The question of how far such a thing gets is as old as the cannon, and Galileo, around 1600, found the idea that cracks it: the flight is not one motion but two, happening at the same time.

**Splitting the kick.** The ball leaves the boot with velocity $\vec{v}$: speed $v_0$ aimed at angle $\alpha$ above the ground. One arrow - but the field asks a horizontal question (how far?) and the goalkeeper a vertical one (how high?). So we split the arrow into a horizontal component $v_x$ and a vertical component $v_y$. Draw the three together and they form a right triangle: $v_0$ is the hypotenuse, and $\alpha$ sits between $v_0$ and $v_x$.

<p><img style="display: block; margin-left: auto; margin-right: auto;" src='https://i.postimg.cc/Vv2KqZqP/footbal-diagram.png' width="500"></p>
<p style="text-align: center;"><em>Fig. 2.2: A kick decomposed: speed $v_0$ at angle $\alpha$ splits into horizontal and vertical components, and the two together draw the arc (source: AI generated).</em></p>


That is exactly the triangle sine and cosine were invented for. In a right triangle, the cosine of an angle is the adjacent side over the hypotenuse, and the sine is the opposite side over the hypotenuse. Multiply each by the hypotenuse $v_0$ and you have the sides themselves:

$$v_x = v_0\cos\alpha, \qquad v_y = v_0\sin\alpha.$$

So $\cos\alpha$ is the share of the kick that travels along the field, and $\sin\alpha$ the share that travels upward. A flat 10 degree kick puts nearly everything into $v_x$; a ball kicked straight up at 90 degrees has $\cos(90^\circ) = 0$ and its whole speed vertical; at 45 degrees the two shares are equal.

Let SageMath do the bookkeeping of that triangle for a 15 m/s kick at 30 degrees:



````sage
# the two components of a 15 m/s kick at 30 degrees
N(15*cos(pi/6))
N(15*sin(pi/6))
````


> **Note: angles in SageMath are in radians.** Your protractor says degrees, but `sin()`, `cos()` and in Sage expect radians. A full turn is $2\pi$ instead of 360, so 180 degrees is `pi` and 30 degrees is `pi/6`. To convert by hand, remember one equation: $\pi\ \mathrm{rad} = 180^\circ$. Dividing both sides by 180 turns degrees into radians, dividing by $\pi$ turns radians back into degrees. SageMath can carry the multiplication factors for you, exactly because $\pi$ stays exact:
>
> ```sage
> # degrees -> radians: multiply by pi/180
> N(30*pi/180)                     # 30 degrees in radians, 0.524
>
> # radians -> degrees: multiply by 180/pi
> N(pi/6*180/pi)                   # pi/6 in degrees, 30.0
> ```
>
> The same factors work inside the trig functions, which beats memorizing special angles: `cos(45*pi/180)` is cos of 45 degrees without knowing that 45 degrees is `pi/4`.


13.0 m/s along the field, 7.5 m/s upward. And because the components are the legs of the triangle with $v_0$ as the hypotenuse, Pythagoras has to reassemble them into the kick speed:




````sage
# the two components reassemble into the original speed
N(sqrt(15*cos(pi/6)^2 + 15*sin(pi/6)^2))   # note: 15*cos(...)^2 means (15*cos(...))^2
````



Fifteen, the speed the kick left with.

**Two motions that ignore each other.** The split only pays off because nature cooperates: the two directions leave each other alone. Moving horizontally does not change how the ball falls, and falling does not push it sideways. So each direction is its own little problem, and each is one you have already solved:

- *Horizontal:* no force acts that way when we ignore air, so the ball coasts at the steady speed $v_x$. Distance is speed times time: $x = v_x\,t$.
- *Vertical:* the ball climbs at $v_y$ while gravity pulls it down with the steady acceleration $g$ from Chapter 1's well. There a stone dropped from rest fell $\tfrac{1}{2}gt^2$ in time $t$; here that same fall is subtracted from a climb.

Put the components in and the whole flight is two functions of time:

$$x(t) = v_0\cos\alpha\, t, \qquad y(t) = v_0\sin\alpha\, t - \tfrac{1}{2}g t^2.$$

The angle $\alpha$ decides how the kick divides, $v_0$ how big it is, and $g$ how hard the planet pulls back. Every projectile question in this book starts from this pair.



## **2.3 Symbolic variables**

So far we stored *numbers* under names. SageMath can also store letters. The command `var()` introduces symbolic variables:




````sage
# symbolic variables have no value yet; they are letters we can compute with
var('t alpha v0 g')
````



Here `t` is time, `alpha` the kick angle, `v0` the initial speed and `g` the gravitational acceleration. They are now empty slots. Any formula built from them stays a formula.

Now the physics of Section 2.2 goes into SageMath. A projectile launched at angle $\alpha$ with speed $v_0$ has, ignoring air resistance, the motion:




````sage
x(t) = v0*cos(alpha)*t
y(t) = v0*sin(alpha)*t - 1/2*g*t^2

show(x(t))
show(y(t))
````



Two functions, defined exactly like the named numbers of Chapter 1, only now every quantity is symbolic. The parentheses in `x(t)` say: the function depends on `t`. The letters `alpha`, `v0` and `g` ride along as parameters.



## **2.4 When does the ball land?**

The ball lands when its height drops back to zero. That is an equation, and SageMath solves equations symbolically with `solve()`:



````sage
solutions = solve(y(t) == 0, t)
solutions
````

````sage
# keep the first solution (index 0) and strip off its right side
t_land = solutions[0].rhs()
show(t_land)
````



Two answers: `t == 0` (the kick itself, obviously true but useless) and the interesting one. Solutions come back as a list, so we can pick the second one by index, and `.rhs()` keeps just the value: the ball is in the air for $2v_0\sin\alpha / g$. Twice the time to the apex, as expected from symmetry.

> **Note:** `solve()` returns a list of *equations*, not values. Indexing with `[0]` picks the first equation, `.rhs()` strips off its right side. If you ever see `t == ...` printed where you wanted a bare expression, you forgot the `.rhs()`.




## **2.5 The range**

The ball lands at horizontal position $x(t_{\mathrm{land}})$. Let SageMath substitute the landing time into the flight function all by itself:



````sage
R(alpha) = x(t_land)
show(R(alpha))
````



It works, but it looks unfinished: sines and cosines fight each other. That is what `simplify()` and its sharper cousin `trig_reduce()` are for:




````sage
show(R(alpha).trig_reduce())
````



There is the famous result, $\sin(2\alpha)$ and all. The range of a projectile is

$$R = \dfrac{v_0^2 \sin 2\alpha}{g}.$$

`trig_reduce()` is one wrench in a whole toolbox. SageMath has a shelf of simplification commands, because "simple" means different things in different situations: sometimes you want a product, sometimes a sum, sometimes one trig function instead of three. A quick tour on small examples:



````sage
var('x')

show((x^2 - 4).factor())                       # factor into a product: (x + 2)*(x - 2)
show(((x + 1)^2 - x^2).expand())               # the opposite: multiply out, 2*x + 1
show((sin(x)^2 + cos(x)^2).trig_simplify())    # pack trig functions together: 1
show(((x - 1)/(x^2 - 1)).simplify_rational())  # cancel fractions: 1/(x + 1)
````


`factor()` pulls a polynomial apart into products and `expand()` flattens it back into a sum, `trig_simplify()` packs trigonometric functions together (the reverse of `trig_reduce()`, which splits them), and `simplify_rational()` cancels fractions. If you do not want to think about which one fits, `simplify_full()` chains several of them in one go, and often, but not always, lands on something readable.

> **Note:** There is no single command that makes a formula "simple", because SageMath cannot know what *you* consider simple: `(x + 2)*(x - 2)` and `x^2 - 4` are both correct, and each is better for a different purpose. Work as we did here: look at the result, decide what bothers you about it, and pick the tool for exactly that.




## **2.6 The optimal angle**


Now the real question: which $\alpha$ makes $R$ largest?

The tool for that is the *derivative*, so a word before the machinery, because the command that computes it does not tell you what one *is*. The derivative of a curve at a point is the slope of its tangent line there: rest a straight ruler on the curve so it touches at exactly one point, tilted exactly as the curve is tilted at that point, and the ruler's rise over run is the value of the derivative. A positive derivative means the curve climbs, a negative one means it falls, and the steeper the tangent, the faster.

Here is a way to *see* that on any curve, not just our range formula. We borrow `plot()` from Chapter 3 ahead of schedule; for now, read the code as "draw these curves". The sine is a good test subject, because it does all three things a curve can do, one after another:



````sage
var('x')

f(x) = sin(x)

# the tangent at an arbitrary point x0: value plus slope times offset
tangent(x, x0) = f(x0) + diff(f(x), x).subs(x = x0)*(x - x0)

# three points: rising, at the maximum, at the minimum
tan_rise  = tangent(x, pi/3)     # derivative cos(pi/3) = 1/2, positive
tan_max   = tangent(x, pi/2)      # derivative cos(pi/2) = 0, horizontal
tan_min   = tangent(x, 3*pi/2)    # derivative cos(3*pi/2) = 0, horizontal

# three tangents
sine_plot = plot(f(x), xmin=0, xmax=2*pi, color = 'blue', legend_label = '$\\sin(x)$')
tan_rise_plot = plot(tan_rise, xmin=pi/3-0.7, xmax=pi/3+0.7, color = 'red', legend_label = 'tangent at x = $\\pi/3$', linestyle='--')
tan_max_plot = plot(tan_max, xmin=pi/2-1, xmax=pi/2+1, color = 'green', legend_label = 'tangent at x = $\\pi/2$ (max)', linestyle='--')
tan_min_plot = plot(tan_min, xmin=3*pi/2-1, xmax=3*pi/2+1, color = 'purple', legend_label = 'tangent at x = $3\\pi/2$ (min)', linestyle='--')

combined_plot = sine_plot + tan_rise_plot + tan_max_plot + tan_min_plot

combined_plot.show(frame=True)
````


Look at the slopes. At $x = \pi/3$ the tangent climbs alongside the sine wave: the curve is on its way up, and the tangent's tilt is the derivative there, positive. At $x = \pi/2$ the sine wave reaches its crest, and its tangent is dead flat: the derivative is zero. At $x = 3\pi/2$ the wave bottoms out in its trough, and the tangent is flat again: the derivative is zero there too. So a zero derivative does not whisper "maximum" - the crest and the trough both have one. The difference is what the curve *does* around the point:

- at a **maximum** the curve rises, flattens, then falls: the derivative changes sign from positive to negative;
- at a **minimum** the curve falls, flattens, then rises: the derivative changes from negative to positive.

That is the method of this section in one picture: **look for the points where the tangent is horizontal, then check which kind they are.**

The same game on the curve we actually care about, with the tangent glued to the maximum this time. The range of a 15 m/s kick on Earth, drawn over the angle:



````sage
var('a')

R_curve(a) = R(a).subs(v0 = 15, g = 9.81)

# the slope of the range curve at 30 degrees
slope30 = diff(R_curve(a), a).subs(a = pi/6)
tangent30(a) = R_curve(pi/6) + slope30*(a - pi/6)

# and the tangent at the suspected maximum, 45 degrees
slope45 = diff(R_curve(a), a).subs(a = pi/4)
tangent45(a) = R_curve(pi/4) + slope45*(a - pi/4)

range_plot   = plot(R_curve(a), xmin = 0, xmax = pi/2, color = 'blue', legend_label = 'range R')
tan30_plot  = plot(tangent30(a), xmin = 0, xmax = pi/2, color = 'red', legend_label = f'tangent at 30 degrees or {round(30*pi/180, 2)} radians', linestyle = '--')
tan45_plot  = plot(tangent45(a), xmin = 0, xmax = pi/2, color = 'green', legend_label = f'tangent at 45 degrees or {round(45*pi/180, 2)} radians', linestyle = '--')

combined_plot = range_plot + tan30_plot + tan45_plot

combined_plot.show(frame = True, axes_labels = ['angle [rad]', 'R [m]'])
````


The red line touches the blue curve at 30 degrees and copies its tilt: the range is still climbing there. The green line rests flat on the top of the hill at 45 degrees: the derivative is zero exactly at the maximum. Slide a tangent along the curve by hand and you can feel the method working - climb, flatten, tip over.

SageMath computes the slope of the tangent for us, with `diff()` ("differentiate", which means "to find derivative"):



````sage
show(diff(R(alpha), alpha))
````


> **Note:** Differentiating by hand follows a small set of rules: powers bring the exponent down and lower it by one, $\sin$ turns into $\cos$, and products or nested functions have their own recipes (the product and chain rules). Our range formula already needs the chain rule, because of the $\sin 2\alpha$ inside. The practical advantage of `diff()` is that SageMath knows all the rules and applies them for us: you do not have to remember a single one to find a maximum, and sometimes the formulas are far too long to differentiate by hand anyway.

Set the derivative to zero and solve, this time for `alpha`:




````sage
optimal = solve(diff(R(alpha), alpha) == 0, alpha)
optimal
````


One family of answers: $\alpha = \pi/4$, that is **45 degrees** (plus full turns and the mirrored kicks, which land backwards). Look at what is *not* in the answer: no $v_0$, no $g$. The optimal angle does not depend on how hard you kick, or on which planet you are standing on.

But a zero derivative only says *the tangent is horizontal there*, and the sine picture above already showed what that costs: the top of a hill and the bottom of a valley both have horizontal tangents. So before celebrating, check which kind of point we found. The tool is the *sign* of the derivative on either side of the point, exactly the sign change we read off the sine wave. Let us look at the slope of the range curve below and above 45 degrees, in two small steps instead of one long line:



````sage
# differentiate once and keep the derivative as its own formula
dR = diff(R(alpha), alpha)
show(dR)
````

````sage
# the slope of the range curve at 30 and at 60 degrees
show(dR.subs(alpha = pi/6))    # positive: the range is still rising at 30 degrees
show(dR.subs(alpha = pi/3))    # negative: the range is already falling at 60 degrees
````


Positive below, negative above: the derivative flips from plus to minus at 45 degrees, and that sign change is the signature of a **maximum**.

A second sanity check needs no derivatives at all: evaluate the range slightly left and slightly right of the suspected top and compare (borrowing the numbers of Section 2.7 for a moment):



````sage
# both neighbors of 45 degrees should land shorter than 45 degrees itself
R_values = R.subs(v0 = 15, g = 9.81)
N(R_values(40*pi/180))    # a flat-ish kick, 40 degrees
````

````sage
N(R_values(45*pi/180))    # the candidate
````

````sage
N(R_values(50*pi/180))    # a high-ish kick, 50 degrees
````


22.6 m, 22.9 m, 22.6 m: both neighbors lose to the candidate in the middle.

And for those who prefer one computation instead of two sign checks, calculus bundles the check into a single number: the **second derivative**. If the first derivative is the slope of the curve, the second is the slope *of the slope*; a maximum sits where the slope decreases through zero, so the second derivative there is negative:



````sage
# differentiate R twice, keep the formula, then substitute 45 degrees
d2R = diff(R(alpha), alpha, 2)
show(d2R.subs(alpha = pi/4))
````


With our formula this is $-4v_0^2/g$, negative for any kick on any planet, so the curve bends downward: $\pi/4$ is the top. And notice the symmetry around it:




````sage
# kicking at 30 degrees reaches exactly as far as kicking at 60 degrees
simplify(R(pi/6) - R(pi/3))
````



Zero. Two angles that add to 90 degrees have the same range, one flat and fast, the other high and slow. Only the 45 degree kick is alone.




## **2.7 Numbers for our football**

Formulas are nice, but the football field has meters on it. Substitute the values with `subs()` and let `N()` do the arithmetic:



````sage
# 15 m/s kick, Earth gravity, 45 degrees
N(R(pi/4).subs(v0 = 15, g = 9.81))
````



About `22.9` m. And while we are here, the flight time and the maximum height of that same kick:




````sage
# how long the ball is in the air
N(t_land.subs(v0 = 15, alpha = pi/4, g = 9.81))
````

````sage
# symmetry: the apex sits at half the flight time
t_apex = t_land/2
N(t_apex.subs(v0 = 15, alpha = pi/4, g = 9.81))
````

````sage
# the height of the ball at that moment
N(y(t_apex).subs(v0 = 15, alpha = pi/4, g = 9.81))
````


The same result is available with the derivative machinery from Section 2.6, only this time with numbers in. The apex is where the tangent on the height curve is horizontal, so first write that condition with our kick substituted, and look at it before solving:



````sage
# the horizontal-tangent condition on the height curve, numbers in
apex_eq = diff(y(t), t).subs(v0 = 15, alpha = pi/4, g = 9.81) == 0
show(apex_eq)
````


A falling straight line set to zero, and its single root is the apex. Now solve it:



````sage
# the apex from the derivative: one equation, one root
t_apex_d = solve(apex_eq, t)[0].rhs()
N(t_apex_d)
````

````sage
# the same apex, so the same height
N(y(t_apex_d).subs(v0 = 15, alpha = pi/4, g = 9.81))
````


Same 1.08 s, same 5.7 m: the symmetry shortcut and the derivative agree. A real goalkeeper would point out that air resistance shortens all of this; we will put the air back in a later chapter.

> **Note:** Watch the order of operations we used, because it is the standard workflow for symbolic physics: *state the model* (two functions), *solve one condition* (landing), *substitute into the other* (range), *differentiate to optimize*, *only at the very end put in numbers*. Numbers went in exactly once, at the end. Compare that with a calculator, where you must commit to numbers in the first line and every later step inherits the rounding.

## **2.8 What you can do now**

- **A projectile splits into components**: $v_x = v_0\cos\alpha$ along the field, $v_y = v_0\sin\alpha$ upward, the legs of a right triangle with the kick speed $v_0$ as the hypotenuse; the two directions then evolve independently, one coasting, one falling.

- **`var()`** introduces symbolic variables; formulas built from them stay formulas and can be manipulated exactly.
- **`solve(equation, unknown)`** returns a list of solutions as equations; pick one with `[i]` and strip the right side with `.rhs()`.
- **`simplify()`, `trig_reduce()`, `factor()`, `expand()`, `trig_simplify()`, `simplify_rational()`, `simplify_full()`** clean up formulas, each in its own way; a clean formula is easier to check against a textbook.
- **`diff(expr, x)`** differentiates; the derivative is the slope of the tangent line, the sine picture showed all three cases at once (rising tangent, flat at the maximum, flat at the minimum), and maxima and minima live where the derivative is zero, confirmed by the sign change of the derivative around the point or by the sign of the second derivative.
- **`.subs(name = value)`** substitutes numbers into formulas; substitute once, at the end, and let `N()` finish the job.
