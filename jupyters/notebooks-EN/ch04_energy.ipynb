## **4.1 The task**

> A cart starts from rest at the top of a **20 m** high rollercoaster hill and rolls down, through a valley and over a **10 m** bump, all frictionless. How fast is it moving at the bottom, at the top of the bump, and where does it fly off if the track suddenly ends?

## **4.2 Energy in two formulas**

A cart of mass $m$ at height $y$ moving with speed $v$ carries



```sage
var('m g y v y0')

E_pot = m*g*y                    # potential energy
E_kin = 1/2*m*v^2                # kinetic energy
E_initial = m*g*y0               # total energy: cart at rest at height y0
```



Energy conservation says their sum never changes, because nothing in this rollercoaster does work on the cart: no friction, no engine. At the start the cart rests at height $y_0$, so the total is $E = mgy_0$, the number the cell above stores as `E_initial`. At any other point of the track:




```sage
# speed as a function of height follows from energy conservation
v_of_y_all = solve(E_kin + E_pot == E_initial, v)
show(v_of_y_all)
```

```sage
# positive root
v_of_y = v_of_y_all[1].rhs()
show(v_of_y)
```


The positive root: $v = \sqrt{2g\,(y_0-y)}$. Look at what dropped out: the mass $m$ is gone. The speed does not depend on how heavy the cart is, same as a hammer and a feather falling in vacuum. And look what stayed a symbol: the starting height $y_0$. The formula serves any hill; which hill exactly is decided by the values, and those go into one list of equations:



```sage
# numerical values of the parameters, collected in one list
values = [y0 == 20,               # m, starting height
          g == 9.81]              # m/s^2, standard gravity
```


> **Note:** The `values` list is a trick worth stealing. In Sage an equation is an object like any other, `y0 == 20` included, so equations can be stored in a list - and `.subs()` happily eats the whole list at once. It does the same job as the keyword form of `.subs()`, but the numbers now live in one named place: when a new parameter shows up, it joins the list, and when the hill gets taller, one line changes and every later `.subs(values)` follows. Each equation carries its unit as a comment, so the list doubles as a little table of the physics input.

Two numbers right away:




```sage
# speed at the bottom of the valley (y = 0)
speed_bottom = v_of_y.subs(y = 0).subs(values)

# in m/s
speed_bottom
```

```sage
# in km/h
speed_bottom*3.6
```

```sage
# speed at the top of the 10 m bump
speed_top = v_of_y.subs(y = 10).subs(values)

# in m/s
speed_top
```

```sage
# in km/h
speed_top*3.6
```


Nearly 70 km/h at the bottom, about 50 km/h over the bump. Notice also the shape: speed depends only on the *height* lost, not on whether the drop was steep or gentle. The valley floor could be tilted, curved, anything; at equal heights, equal speeds.




## **4.3 Three energies in one plot**

To see the whole story at once, plot potential, kinetic and total energy along the track. The track itself we describe by a height profile $y(x)$, any smooth wavy function with a 20 m start, a valley and a 10 m bump. Here is one, built from a cosine:



```sage
var('x')

# a made-up track: starts at 20 m, dips to 5 m, bumps at 10 m
y_track(x) = piecewise([((0, 20), 7.5*cos(2*pi*x/40) + 12.5),
                        ((20, 80), 2.5*cos(2*pi*x/40) + 7.5)])

p_track = plot(y_track(x), xmin=0, xmax=80, ymin=0, ymax=25, axes_labels = ['$h\\;\\mathrm{[m]}$', '$x\\;\\mathrm{[m]}$'], 
               color='green', frame=True, gridlines=True)
p_track
```


The new tool here is `piecewise`. It takes a list of pairs, each pair being an interval and the formula valid on it. Asked for a value, it checks which interval the input falls into and evaluates that formula. Our track needs exactly this: one cosine carrying the cart down to the valley (0 to 20 m) and a flatter one over the bump (20 to 80 m), two stretches stitched into one piecewise-defined function, for each interval separately. Using proper mathematical notation the piecewise-defined function would look like this:

$$
y_{\mathrm{track}}(x) =\left\{ \begin{array}{ll}
7.5\cos\left(\dfrac{2\pi x}{40}\right) + 12.5, & 0 \le x \le 20 \\
2.5\cos\left(\dfrac{2\pi x}{40}\right) + 7.5, & 20 < x \le 80
\end{array} \right.
$$


Height alone, as a picture. Now the energies, and the height profile is all the input they need:




```sage
E_track(x) = E_pot.subs(y = y_track(x))          # potential along the track
E_k_track(x) = E_initial - E_track(x)            # kinetic = total - potential
```



Kinetic as "total minus potential" is energy conservation a second time, and it is worth slowing down here. The total $E = E_k + E_p$ is the same number at every point of the track, $E = mgy_0$ from the resting start. So once the potential $E_p(x) = mg\,y(x)$ is known at a position $x$, the kinetic energy is simply what remains of the budget: $E_k(x) = E - E_p(x)$. The code line `E_k_track(x) = E_initial - E_track(x)` is that one sentence, written in Sage.

Notice what we never touch: velocity. On a curved track, computing $v$ at each $x$ directly would be genuinely unpleasant bookkeeping. Velocity $\vec{v}$ is a vector: as the cart rolls, its direction turns with the slope, and both components change continuously, $v_x$ and $v_y$ each doing their own dance along the track. The speed would have to be assembled afterwards from the components, $v = \sqrt{v_x^2 + v_y^2}$, which first requires both of them as functions of $x$ and nobody wants that algebra. Energies escape all of this, because they are scalars: $E_p = mgy$ and $E_k = \tfrac{1}{2}mv^2$ are single numbers, no direction attached, and they add and subtract like numbers. That is the quiet power of the energy method - it trades a vector problem for a scalar subtraction. Now all three in one picture. The mass canceled in section 4.2, so it never needed a number; now it does, and it joins the list:




```sage
# the mass of the cart joins the parameter list
values += [m == 500]              # kg

p1 = plot(E_track(x).subs(values), xmin = 0, xmax = 80, color = 'blue',
          legend_label = 'potential energy')
p2 = plot(E_k_track(x).subs(values), xmin = 0, xmax = 80, color = 'red',
          legend_label = 'kinetic energy')
p3 = plot(E_initial.subs(values), xmin = 0, xmax = 80, color = 'black', linestyle = '--',
          legend_label = 'total energy')

p_format = plot([], '$x\\;\\mathrm{[m]}$', '$E\\;\\mathrm{[J]}$', ymax=120e3, frame=True, gridlines=True)

p_energies = p1 + p2 + p3 + p_format

show(p_energies)
```


> **Note:** The `p_format` line is a small trick. `plot([])` draws an empty list of curves, so it paints nothing, but it still accepts every formatting option a real plot does. When plots are combined with `+`, the options of all summands are merged, so the empty plot carries the frame, gridlines, axis labels and `ymax` into the final picture. The same options could sit inside any of `p1`, `p2` or `p3` instead and the result would look identical; keeping them in a separate empty plot is just a little more flexible when the combination of plots changes. Drop one energy, add a fourth curve, show `p1` and `p2` alone, and the formatting still travels with `p_format`, no retyping into whichever plot happens to remain.

> **Note:** One line, `values += [m == 500]`, and the mass is part of the story. Adding two lists with `+` glues them into one longer list, and `+=` extends `values` in place, so every later `.subs(values)` sees the mass too. Prefer to keep the old list untouched? `values_updated = values + [m == 500]` builds a new list and leaves `values` alone. Python readers may know `.append()`: it also adds a single element to the end, in place - `values += [...]` does the same job by plain list addition, with the bonus that it can take several equations at once, `values += [m == 500, r == 15]`.



```sage
# plot with track and energies one above another using graphics array
# figsize parameter controls dimensions of the output plot
graphics_array([[p_track], [p_energies]]).show(figsize=[10,7])
```



Read the picture: the blue potential wave and the red kinetic wave are mirror images, trading height for speed and back, while the dashed black total sits perfectly flat. That flat line *is* the law of conservation of energy, drawn. Where the track dips, blue is low, red is high: the cart is low and fast. Over the bump, they meet halfway.

*Fig. 4.2: Three energies, one law: the red kinetic and blue potential waves trade exactly, while the dashed total line sits flat.*

> **Note:** The kinetic energy must never go negative, $E_k = E - mg\,y(x) \ge 0$, so the cart can never climb above its starting height, $y = 20~\mathrm{m}$. If someone builds a 25 m hill into this track, the red curve would dip below zero and the picture would be lying. The equation, not the drawing, decides what is physical.




## **4.4 Flying off the end**

Now the track ends abruptly at the top of the 10 m bump, with the cart moving horizontally at the speed we computed there. From Chapter 2 we know everything about a launch at a horizontal speed: it is a projectile with angle zero. The flight functions, with the launch speed from energy:



```sage
# horizontal launch from height 10 m at the bump-top speed
v_launch = N(v_of_y.subs(y = 10).subs(values))
v_launch
```

```sage
var('t')

x_flight(t) = 80 + v_launch*t
y_flight(t) = 10 - 1/2*g*t^2

# landing: y_flight == 0, solved symbolically
t_land = solve(y_flight(t) == 0, t)[1].rhs()
N(t_land.subs(values))
```

```sage
# where it lands
N(x_flight(t_land).subs(values))
```



About 1.43 s in the air, landing roughly 20 m past the end of the track. And the landing speed? Energy again, no need for components at all:




```sage
# at landing y = 0: the whole initial budget is kinetic
N(solve(1/2*m*v^2 == E_initial, v)[1].rhs().subs(values))
```



The same 19.8 m/s as at the bottom of the valley: what goes down comes back with the same speed, height for height. Energy bookkeeping makes even the landing a one-liner.

By how much do the bottom speed and the bump-top speed drop? Redraw the three-energy picture with the total line sloping downward. (Hint: friction takes its cut from the *distance travelled*, not from the height, so the loss is a function of $x$. It is one line in Sage, thanks to the named total from section 4.2:



```sage
# energy lost to friction: 30 % of E_initial per 40 m travelled, linear in x
E_loss(x) = 0.3*E_initial/40*x

# the shrinking budget: total energy still available on the track
E_total(x) = E_initial - E_loss(x)

# kinetic = remaining budget minus potential, as before
E_k_track(x) = E_total(x) - E_track(x)
```

## **4.5 What you can do now**

- **Energy conservation** turns any "how fast at height y" question into one equation, $E_k + E_p = \mathrm{const}$; the mass often cancels, so the answer survives even without knowing it.
- **A made-up track function** like `y_track(x)` is a legitimate physics model in the form of a piecewise-defined function where different condition apply on different intervals.
- **A list of equations** like `values = [y0 == 20, g == 9.81]` collects the numerical parameters in one place; handing the whole list to `.subs()` turns every formula numeric at once, and `values += [m == 500]` grows the list as new parameters enter the story.
- **Three plots in one picture** (potential, kinetic, total, using graphics array) makes a conservation law visible.
