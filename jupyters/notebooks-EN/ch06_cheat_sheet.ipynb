| Code | What it does |

```sage
# the exactness rules, in one cell each
1/3 + 1/6      # exact: 1/2
0.33 + 0.17    # approximate: one decimal point spoils the whole result
```


> **Note: forcing exact or approximate.** For a decimal answer without typing `N()`, write one input as a decimal: `10.0` instead of `10`. For exact arithmetic from naturally decimal values, write them as fractions: `981/100` instead of `9.81` - and the trick must cover *all* decimals entering the calculation, one decimal anywhere spoils the exactness. Numbers use a **decimal point, never a comma**. Stay exact as long as possible, go decimal at the end, and round only at the very end, never in the middle.



## **6.3 Variables and functions**

| Code | What it does |
|---|---|
| `g = 9.81` | store a number under a name; unit goes in a comment: `# m/s^2` |
| `var('t alpha v0 g')` | declare symbolic variables: letters, empty slots |
| `t_h(h) = sqrt(2*h/g)` | store a formula as a symbolic function of `h` |
| `t_hg(h, g) = sqrt(2*h/g)` | a function of several variables |
| `t_minutes = t_h/60` | functions are values: scale, add, combine them |
| `show(t_h)` | typeset the formula |



```sage
# the workflow from Chapter 2: state the model, solve, substitute, numbers last
var('t alpha v0 g')

x(t) = v0*cos(alpha)*t
y(t) = v0*sin(alpha)*t - 1/2*g*t^2     # the model, two functions of time

solutions = solve(y(t) == 0, t)        # an equation, solved for t
t_land = solutions[1].rhs()            # pick by index, strip the right side
show(t_land)

R(alpha) = x(t_land)                   # substitute one formula into another
show(R(alpha).trig_reduce())
```


> **Note:** `==` (two equal signs) builds an equation, `=` (one) stores a value. `solve()` returns a list of *equations*, not values: index with `[i]`, strip the value with `.rhs()`. A function and a symbol cannot share a letter (`x(t)` and `x` collide); the fix is a trailing underscore in the code, `x_(t) = ...`, which `show()` typesets as plain $x$ anyway.



## **6.4 Substitution**

| Code | What it does |
|---|---|
| `expr.subs(alpha = pi/4)` | substitute by keyword, one or several at once |
| `values = [y0 == 20, g == 9.81]` | collect parameter values as a list of equations |
| `expr.subs(values)` | substitute the whole list at once |
| `values += [m == 500]` | grow the list in place; several equations at once allowed |
| `values_updated = values + [r == 15]` | the same, but leaving the old list untouched |

Equations are objects like any other, so they live in lists, and `.subs()` eats the whole list in one call. Each equation carries its unit as a comment, so `values` doubles as the table of the physics input.



## **6.5 Solving and calculus**

| Code | What it does |
|---|---|
| `solve(equation, unknown)` | symbolic solution; returns a list of equations |
| `solve([eq1, eq2, eq3], x, y, z)` | a whole system of equations |
| `solutions[1].rhs()` | pick the second solution, keep its right side |
| `find_root(T(h) == 3.2, 1, 100)` | numerical hunter: a root inside the given interval |
| `diff(expr, x)` | the derivative of `expr` with respect to `x` |
| `diff(expr, x, 2)` | the second derivative |
| `solve(diff(R(a), a) == 0, a)` | maximum candidates: where the tangent is horizontal |



```sage
# optimizing, the Chapter 2 recipe
dR = diff(R(alpha), alpha)             # the derivative as its own formula
optimal = solve(dR == 0, alpha)        # horizontal tangents

show(dR.subs(alpha = pi/6))           # positive below the candidate: still rising
show(dR.subs(alpha = pi/3))           # negative above: already falling - a maximum
show(diff(R(alpha), alpha, 2).subs(alpha = pi/4))   # second derivative < 0 says the same
```


> **Note:** A zero derivative means *horizontal tangent*, not *maximum*: crests and troughs both have one. Confirm the kind of point by the sign change of the derivative around it, by the sign of the second derivative, or by evaluating the neighbors (`N(R(40*pi/180))` vs. `N(R(45*pi/180))` vs. `N(R(50*pi/180))`). Use `find_root` when `solve` gives up, e.g. on a mix of a square root and a linear term; it wants the equation written with `==`.



## **6.6 Simplification**

| Code | What it does |
|---|---|
| `simplify(expr)` | the general cleaner |
| `expr.factor()` | pull a polynomial apart into a product |
| `expr.expand()` | the opposite: multiply out into a sum |
| `expr.trig_reduce()` | rewrite trig products as single sines, e.g. $\sin(2\alpha)$ |
| `expr.trig_simplify()` | the reverse: pack trig functions together |
| `expr.simplify_rational()` | cancel fractions |
| `expr.simplify_full()` | several of the above chained in one go |

There is no command that makes a formula "simple", because Sage cannot know what *you* consider simple. Work as in Chapter 2: look at the result, decide what bothers you, pick the tool for exactly that.



## **6.7 Trigonometry and angles**

| Code | What it does |
|---|---|
| `sin(x)`, `cos(x)`, `tan(x)` | trig functions, **radians in, radians out** |
| `30*pi/180` | degrees to radians: multiply by $\pi/180$ |

## **6.8 Lists**

| Code | What it does |
|---|---|

## **6.9 Python in one place**

| Code | What it does |
|---|---|
| `def f(x, y): return x + y` | name a recipe: parameters in, one value out |

| `lambda a: N(...)` | the same, squeezed into one line; no equation, no `show()` |
| `zip(xs, ys)` | join two lists into pairs, like a zipper |

| `"text"` or `'text'` | a string; Python (and SageMath) accepts both quote marks, but a string must open and close with the same one |

```sage
# homemade functions, the Chapter 5 pattern
def y_errorbar(x, y, dy):
    return line([(x, y - dy), (x, y + dy)], color='lime', thickness=1.5)

y_errorbar(0.06, 15.2, 1.2)     # a call computes the line and hands it back
```


> **Note:** `def` for a body of one or more indented lines, `lambda` when it fits on one line. `return` is what makes the answer a value: storable, listable, summable. A `lambda` is pure computing - nothing to solve or differentiate - while a symbolic function like `t_h(h) = ...` is a formula Sage can solve, substitute into and differentiate.

## **6.14 The standard workflow**

1. **Store what is known.** `g = 9.81  # m/s^2`
2. **Declare symbols.** `var('t alpha v0 g')`
3. **State the model.** `y(t) = v0*sin(alpha)*t - 1/2*g*t^2`
4. **Solve conditions symbolically.** `solve(y(t) == 0, t)`, pick with `[i]`, strip with `.rhs()`
5. **Substitute formulas into each other.** `R(alpha) = x(t_land)`
6. **Clean up.** `.trig_reduce()`, `.factor()`, `simplify_full()`
7. **Optimize.** `solve(diff(R, alpha) == 0, alpha)`, confirm with the sign change
8. **Numbers last, in one list.** `values = [v0 == 15, g == 9.81]`, then `.subs(values)`
9. **Go decimal.** `N(...)`, `round(..., 2)`
10. **Draw and save.** `plot(...) + plot_format`, `.save("name.png", dpi=300)`

Never type a long formula, run it, and hope: one small step, check the result, take the next. Sage carries the arithmetic flawlessly; what to compute, and whether the answer is sensible, stays your job.
