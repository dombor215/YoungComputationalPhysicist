## **0.3 Prvá ochutnávka: Sage ako asistent fyzika**

> ***Poznámka:** Nič sa nedeje, ak vám ešte nie je jasné všetko z tohto oddielu. So zmyslom pre pravidelnosti a vzorce (a trochou angličtiny) môžete odhaliť, čo jednotlivé príkazy robia. Skúšajte kód drobne upravovať a sledujte, čo sa zmení. Cieľom tejto kapitoly je ukázať, čo Sage dokáže; všetko podrobne si vysvetlíme v nasledujúcich kapitolách.*

Najobľúbenejší trik systému Sage ho odlišuje od akejkoľvek vreckovej kalkulačky: so vzorcami zaobchádza tak ako papier. Uložme druhý Newtonov zákon – nie čísla, ale samotný *zákon*:



```sage
# typické úlohy v Sage: príklad druhého Newtonovho zákona
var("F m a")                    # deklarácia symbolických premenných
newtons_second_law = a == F/m  # uloženie zákona ako rovnice (dvojitý znak rovnosti)
```

```sage
# pekné zobrazenie uloženého zákona
show(newtons_second_law)
```

```sage
# vyriešime rovnicu pre neznámu m (hmotnosť)
show(solve(newtons_second_law, m))
```


Jedna uložená rovnica, ľubovoľná neznáma – a postup sa dá bez námahy rozšíriť. Sústava troch lineárnych rovníc s tromi neznámymi $x, y, z$:

$$\begin{aligned}
2x + y - z = 3 \\
-x + 3y + 2z = 7 \\
4x - 2y + z = 1
\end{aligned}$$



```sage
# vyriešime sústavu, štyri riadky kódu
var('x y z')
eq1 = 2*x + y - z == 3
eq2 = -x + 3*y + 2*z == 7
eq3 = 4*x - 2*y + z == 1
show(solve([eq1, eq2, eq3], x, y, z))
```


Alebo sa vráťme k fyzike: pohybová rovnica telesa vrhnutého zvisle nahor z výšky $h_0$ začiatočnou rýchlosťou $v_0$,

$$h = h_0 + v_0 t - \frac{1}{2}g t^2,$$

a nechajme Sage nájsť všeobecný vzorec pre časy, v ktorých teleso dosiahne danú výšku $h$ – ručná úprava by zabrala pol strany:



```sage
# zadáme pohybovú rovnicu
var("h h0 v0 t g")
equation = h == h0 + v0*t - 1/2*g*t^2
show(equation)
```

```sage
# všeobecné riešenie pre čas t
show(solve(equation, t))
```

```sage
# dosadíme číselné hodnoty do druhého riešenia
values = [h0==100, v0==10, g==9.81, h==0]
show(solve(equation, t)[1])
show(solve(equation, t)[1].subs(values))
```


Sage vie aj upratať po vás, a to pri výrazoch, na ktoré by vreckové kalkulačky nestačili. Zjednodušme napríklad výraz $\sqrt{\dfrac{m^2 v^4}{m^2 v^2} \cdot x}$:



```sage
# zadáme výraz
var("m v x")
expression = sqrt((m^2 * v^4)/(m^2 * v^2)*x)
show(expression)
```

```sage
# výraz zjednodušený ešte viac
show(expression.canonicalize_radical())
```


Nejde o nacvičovanie písania. Každý z týchto krokov – uložiť vzorec, vyjadriť hľadanú veličinu, dosadiť čísla, zjednodušiť – robí pracujúci fyzik denne. Na papieri by každý z nich zabral minúty a každá minúta by bola príležitosťou na chybu. V Sage tieto mechanické kroky prebehnú bezchybne a nie je potrebné ich znovu kontrolovať. Vám zostáva to podstatné – či fyzika, ktorú ste zadali, je tá, ktorú ste mali na mysli – a práve to si zaslúži vašu pozornosť.



## **0.4 Grafy ako plnohodnotný výstup**

Zadajte funkciu a Sage ju presne a okamžite vykreslí:



```sage
# zadáme funkciu
f(x) = 1/x*sin(x)
show(f)
```

```sage
# jednoduchý graf funkcie f pre x od -30 do 30
show(plot(f, (x,-30,30)))
```


> ***Všimnite si:** Hoci sme formálne zadali $f(x) = \dfrac{1}{x} \cdot \sin x$, Sage ju automaticky zjednodušil na $f(x) = \dfrac{\sin x}{x}$. Dokáže to v oveľa väčšom rozsahu aj pri podstatne zložitejších výrazoch.*

Rovnaký graf v „profesionálnej“ úprave, vhodný do laboratórneho protokolu alebo dokonca odborného vedeckého článku:



```sage
# predchádzajúci graf upravený ako obrázok do publikácie
plot_f_formatted = plot(f, (x,-30,30), color="magenta")
# farby sa zadávajú anglicky, napr. "magenta", "red", "green", ...
show(plot_f_formatted,
     title=r"Plot of the function f = \dfrac{\sin x}{x}",
     frame=True, gridlines=True)
```


Sage prestáva byť kalkulačkou a stáva sa nástrojom fyzika pri porovnávaní: viacero variantov jedného javu v jednom obrázku, s legendou. Tu je tlmený kmitavý pohyb, $x(t) = A\,e^{-bt}\cos(\omega t)$, pri štyroch rôznych nastaveniach amplitúdy, útlmu a frekvencie:





```sage
# definícia tlmených kmitov v závislosti od času t
# a od parametrov A, b, omega
x(t, A, b, omega) = A * e^(-b*t) * cos(omega*t)
show(x)
```

```sage
# krivka 1: základné parametre (A=1, b=1, omega=1)
plot_1 = plot(x(t, 1, 1, 1), (t, 0, 10),
              legend_label=r"$A, b, \omega$ = 1, 1, 1")

# krivka 2: zvýšená frekvencia (omega=5)
plot_2 = plot(x(t, 1, 1, 5), (t, 0, 10),
              color="red",
              legend_label=r"$A, b, \omega$ = 1, 1, 5")

# krivka 3: znížený útlm (b=0.3)
plot_3 = plot(x(t, 1, 0.3, 1), (t, 0, 10),
              color="green",
              legend_label=r"$A, b, \omega$ = 1, 0.3, 1")

# krivka 4: zväčšená amplitúda (A=2)
plot_4 = plot(x(t, 2, 1, 1), (t, 0, 10),
              color="cyan",
              legend_label=r"$A, b, \omega$ = 2, 1, 1",
              frame=True, gridlines=True,
              title="Damped oscillation for various parameters",
              axes_labels=[r"$t$", r"$x(t)$"])
```

```sage
# spojíme všetky grafy do jedného
plot_final = plot_1 + plot_2 + plot_3 + plot_4

# zobrazíme výsledný graf
show(plot_final)
```

```sage
# uložíme graf vo vysokom rozlíšení (300 DPI) ako súbor PNG
plot_final.save("damped_oscillations.png", dpi=300)
```

## **0.5 Prečo Sage a prečo práve v tejto knihe?**

Najprv úprimne o konkurencii. Papier, pero a kalkulačka si na kvadratickú rovnicu úplne postačia. Tabuľkový procesor (napr. MS Excel) nakreslí graf, ak si najprv ručne vypočítate tabuľku jeho bodov. GeoGebra kreslí pekné grafy, ale každá zmena parametra znamená ďalšie klikanie.

Potom práca narastie. Dáta zo skutočného experimentu (kapitola 5) prídu ako tabuľka so stovkou meraní s priloženými neistotami a žiadna kalkulačka na to nemá tlačidlo. Načítať ich do tabuľky, skontrolovať, či v nich nechýbajú hodnoty a či neobsahujú preklepy, zobraziť ich s chybovými úsečkami, preložiť nimi model a z fitu vyčítať fyzikálny obsah – to všetko je náplňou kapitoly 5 a sú to úlohy, kde sa ceruzka už nevráti. Práve takéto výpočty vypĺňajú dni súčasnej fyziky a práve takéto výpočty počítače zvládajú výborne.

SageMath je naším prostriedkom na túto cestu, pretože začiatočníka stretne tam, kde sa práve nachádza:

* **číta sa ako matematika.** `sqrt(2*h/g)` je samotný vzorec, nie program, ktorý je okolo neho obalený.
* **je presný, keď chcete presnosť,** a numerický, keď o to požiadate – a pozná medzi nimi rozdiel (venuje sa tomu kapitola 1).
* **je otvorený a bezplatný:** spustíte ho v ktoromkoľvek prehliadači cez vyššie uvedené odkazy alebo si ho môžete nainštalovať bez akýchkoľvek poplatkov.
* **neskrýva cestu pred vami.** Sage je postavený na Pythone, jazyku vedeckých výpočtov, takže to, čo začína ako superkalkulačka s triviálnym kódom, prerastie – bez zmeny nástroja – do zručností, na ktorých stojí skutočný výskum.

Jedno úprimné varovanie ešte pred prvou kapitolou: Sage za vás myslieť nebude. Aritmetiku zvládne bezchybne, ale čo počítať a či je výsledok zmysluplný, zostáva na vás. Kapitola 1 začína presne tu: kameň, studňa a prečo je počet uvádzaných platných číslic fyzikálnym rozhodnutím, a nie záležitosťou kalkulačky.
