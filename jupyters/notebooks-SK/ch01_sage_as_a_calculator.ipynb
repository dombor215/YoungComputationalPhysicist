## **1.1 Voľný pád jedným vzorcom**

Keď pustíte kameň, nezostane vo vzduchu visieť: padá čoraz rýchlejšie, čím dlhšie padá. Zákonitosť odvodil Galileo pred viac ako 400 rokmi a je pozoruhodne jednoduchá. Ak zanedbáme odpor vzduchu, teleso pustené z pokoja prejde za čas $t$ dráhu (výšku)

$$h = \frac{1}{2}\, g\, t^2,$$

kde $g = 9{,}81\ \mathrm{m/s^2}$ je gravitačné zrýchlenie pri povrchu Zeme (mierne sa mení s nadmorskou výškou a zemepisnou šírkou). Úpravou vzorca dostaneme čas potrebný na pád z výšky $h$:

$$t = \sqrt{\frac{2h}{g}}.$$

Zdvojnásobenie hĺbky pádu neznamená zdvojnásobenie času: čas rastie ako druhá odmocnina, $t \propto \sqrt{h}$. Kameň padajúci z 90 m padá len asi 1,4-krát dlhšie ako kameň padajúci zo 45 m.

### **1.2.1 Súhrn: aritmetika v Sage**

Pred prechodom k úlohe o studni si v krátkosti zhrňme všetko, čo budete pri počítaní v Sage potrebovať. Tento prehľad môžete použiť aj neskôr ako ťahák.

**Základné operácie.** Používajú sa znaky `+`, `-`, `*` a `/`. Delenie `/` Sage chápe ako zlomok a automaticky ho skracuje na základný tvar:



```sage
2/64 + 5/32
```

```sage
18.4 - 3*7
```


Poradie operácií je rovnaké ako na papieri. Na zoskupovanie používajte **stále okrúhle zátvorky** `(` `)`. Hranaté `[ ]` a kučeravé `{ }` zátvorky majú v Sage iný význam (zoznamy, množiny), takže vnorené zátvorky z papiera treba prepísať na okrúhle. Skúste vyčísliť:

$$\dfrac{\left(1.5 + \frac{2}{3}\right) \cdot \left[4.75 - \left(2.25 + \frac{1}{2}\right)\right]}{5.6 + \frac{7.2}{3.4 - \frac{3}{4}}}$$



```sage
# priestor na výpočet (správny výsledok je približne 0.521022383545069)

```


**Celočíselné delenie a zvyšok.** Okrem `/` existujú dva užitočné operátory: `//` dá celú časť podielu a `%` zvyšok po delení:



```sage
17//5, 17 % 5
```


Výsledok `(3, 2)` znamená $17 = 3\cdot 5 + 2$. Pri iných výpočtoch uvidíte aj `factorial(5)` ($5! = 120$) a `abs(-3)` (absolútna hodnota).

**Presné a približné čísla.** Približnú hodnotu získate pomocou `N(výraz)`, resp. `(výraz).n()`, a počet platných číslic určíte cez `digits=`. Funkcie `RR()` a `QQ()` menia číslo na reálne, resp. racionálne:



```sage
N(22/7 + 1, digits=3), RR(22/7 + 1), QQ(-1.25)
```


Posledná hodnota sa zobrazí ako zlomok `-5/4`. Pri iracionálnych číslach (napr. $\sqrt{2}$) `QQ()` samozrejme nepomôže, pretože ich nemožno zapísať zlomkom.

**Matematické konštanty.** Ludolfovo číslo $\pi$ sa zapisuje `pi` a Eulerovo číslo $e$ sa zapisuje `e`. V Sage sa správajú **ako symboly**: zostávajú v presnom tvare, kým si nevyžiadate približnú hodnotu. Sage ich vie vypočítať s ľubovoľnou presnosťou:



```sage
(3.5*pi + 12*8)/5
```

```sage
N(pi, digits=100)
```


**Mocniny a odmocniny.** Mocnina sa píše `^` alebo `**`. Sage zvláda aj obrovské čísla, ktoré by kalkulačka neunesla (skúste `2^10000`). Druhá odmocnina je `sqrt()` a odmocniny vyšších rádov zapíšeme cez zlomkový exponent, $\sqrt[n]{x} = x^{1/n}$:



```sage
sqrt(256), 256^(1/4)
```


Odmocnina zostáva presná, kým nepoužijete `N()`. Vyskúšajte `sqrt(2)` a `N(sqrt(2))`. Pozor: exponent `1/4` musí byť v zátvorke, inak by `256^1/4` znamenalo $(256^1)/4$.

**Elementárne funkcie.** Prehľad zápisov:

| Funkcia | Zápis v Sage |
|---|---|
| $\sin x$, $\cos x$, $\operatorname{tg} x$ | `sin(x)`, `cos(x)`, `tan(x)` |
| $\sin^{-1} x$, $\cos^{-1} x$, $\operatorname{tg}^{-1} x$ | `arcsin(x)`, `arccos(x)`, `arctan(x)` |
| $x^n$ | `x^n` alebo `x**n` |
| $\sqrt{x}$ | `sqrt(x)` |
| $\sqrt[n]{x}$ | `x^(1/n)` |
| $e^x$ | `exp(x)` alebo `e^x` |
| $\ln x$ | `log(x)` |
| $\log_a x$ | `log(x, a)` |
| $\log x$ (desiatkový) | `log(x, 10)` |

Argumentom trigonometrických funkcií je uhol **v radiánoch**. Sage vracia presné hodnoty, nie zaokrúhlené:



```sage
sin(pi/4)
```


To posledné číslo udáva, o koľko dlhšie trvá pád z 20 m než pád z 10 m. Všimnite si meno, ktoré sme funkcii dali: `t_h`, nie `t`. Má to svoj dôvod a je to malé poučenie o poriadku v notebooku.

> **Poznámka: mená sa prepisujú.** Niekde vyššie sa pod menom `t` uložilo číslo `3.02891266407691`. Keby sme teraz napísali `t(h) = sqrt(2*h/g)`, toto číslo by zmizlo a nahradila by ho funkcia; skoršie bunky používajúce `t` by po opätovnom spustení potichu znamenali niečo iné. Keď sa veličina zmení *druhom* (z čísla na funkciu), zmeňte aj jej meno: `t_h(h)` pre $t(h)$, `t_hg(h, g)` pre $t(h, g)$. Vhodné by bolo aj `time_in_minutes`, ale často používané veličiny si zaslúžia stručné a čitateľné mená.

Funkcia `t_h(h)` drží `g` odkazom, nie hodnotou. Ak neskôr zmeníte `g` a znova spustíte definíciu funkcie, použije sa nové `g`; ak `g` zmeníte a definíciu *nespustíte* znova, ostáva v nej zapísaná stará hodnota. A funkcia môže mať aj viac argumentov:



```sage
t_hg(h, g) = sqrt(2*h/g)   # t ako funkcia dvoch premenných
t_hg(45, 10).n()           # .n() na konci funguje ako N()
```


Keď je `g = 1.62` (Mesiac) k dispozícii ako argument, tento jediný riadok zodpovie aj cvičenie 3. Funkcie možno upravovať aj ako celok:



```sage
t_minutes = t_h/60         # tie isté časy v minútach
show(t_minutes)
```

## **1.6 Mnoho hĺbok naraz: zoznamy**

Doteraz každý príkaz vytvoril jedno číslo. Vo fyzike sa málokedy pracuje s jedným číslom. Predpokladajme, že chcete poznať čas pádu pre šesť hĺbok naraz. Najprv hĺbky uložíme do **zoznamu** (*list*):



```sage
# zoznam hĺbok, pre ktoré chceme poznať časy
depths = [5, 10, 15, 20, 25, 30]
results = [t_h(h).n() for h in depths]
results
```


Zoznam sa vytvára pomocou hranatých zátvoriek `[` `]` a jeho položky sa oddeľujú čiarkami – ďalší dôvod, prečo sa v číslach používa desatinná bodka, nie čiarka. Riadok `[t_h(h).n() for h in depths]` sa číta doslova takto: *pre každé `h` zo zoznamu `depths` vypočítaj `t_h(h).n()` a zhromaždi výsledky do nového zoznamu s názvom `results`.* Kľúčové slová `for` a `in` patria do takzvaného **cyklu for**. Tento stručný zápis sa nazýva *list comprehension* (zostavenie zoznamu), elegantná pythonovská konštrukcia. Ak už trochu poznáte Python, tu je ten istý výpočet zapísaný tradičným spôsobom:



```sage
depths = [5, 10, 15, 20, 25, 30]    # zoznam hodnôt, cez ktoré prechádzame
results = []                        # pripravený prázdny zoznam

for h in depths:                    # tradičný cyklus for
    value = t_h(h).n()
    results.append(value)           # .append() pripojí hodnotu do zoznamu

results
```


Jedno upozornenie, rovnaké poučenie ako pri `t`: v týchto bunkách už `h` nie je hĺbka našej studne. Definíciou `t_h(h)` sa `h` aj `g` zmenili na *symbolické premenné*. Zostavenie zoznamu potom použilo `h` ako bežiace meno. Číslo 45 prežíva vo funkcii len ako hodnota, ktorú jej zadáte. Nič sa nestratilo, ale mená sa posunuli; preto nové veličiny dostávajú nové mená.

Pracovať so zoznamom znamená vyberať z neho hodnoty späť, a to sa robí pomocou **indexov**:



```sage
results[0]      # prvá hodnota v zozname
```

```sage
last_value = results[-1]   # posledná hodnota v zozname
last_value
```


Indexy sa počítajú takto: 1. hodnota má index `0`, 2. hodnota index `1`, 3. hodnota index `2` atď. Python (a teda aj SageMath) začína počítať od `0`, čo spočiatku vyzerá čudne, ale vypláca sa pri záporných indexoch: posledná hodnota má index `-1`, predposledná `-2`, takže na dosiahnutie konca zoznamu nepotrebujete poznať jeho dĺžku. Šiesta hodnota zoznamu `results` má index `5` – a áno, v Pythone povieme „šiesta hodnota má index päť“ bez mihnutia oka.

Hodnoty zo zoznamu možno v ďalších výpočtoch používať ako akékoľvek iné čísla:



```sage
results[2]/60     # tretia hodnota prevedená na minúty
```


A zoznamy sa bez námahy rozrastajú. Tridsať výpočtov nie je problém:



```sage
depths30 = [5, 6 .. 30]   # rozsah: od 5 do 30 s krokom 1 (krok určuje druhá hodnota)
results30 = [t_h(h).n() for h in depths30]
results30[5:10]          # 6. až 10. hodnota; v reze sa posledný index nezahŕňa
```

```sage
results30[10]
```


Dlhé zoznamy ako `[5, 6 .. 30]` sa zapisujú pomocou dvoch bodiek medzi prvými dvoma hodnotami; krok sa určí z rozdielu medzi nimi. Rez (*slice*) `results30[5:10]` je zoznam vystrihnutý z iného zoznamu a pravý index sa opäť nezahŕňa: dostaneme prvky s indexmi 5, 6, 7, 8, 9.
