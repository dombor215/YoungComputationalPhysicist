## **2.2 Fyzika letiacej lopty**

Skôr než dáme SageMath niečo na spracovanie, uveďme slovami fyziku kopu, pretože každý vzorec v tejto kapitole vyrastá z jednej starej myšlienky.

Kopnutá lopta je *teleso vrhnuté v gravitačnom poli* (projektil): teleso, ktoré dostalo jeden impulz a potom letí samo, bez toho, aby ho čokoľvek riadilo. Delové gule, basketbalové lopty aj voda z fontány sú takisto takéto telesá. Otázka, ako ďaleko sa také teleso dostane, je rovnako stará ako delo a Galileo okolo roku 1600 objavil myšlienku, ktorá ju rozlúskne: let nie je jeden pohyb, ale dva pohyby, ktoré prebiehajú súčasne.

**Rozklad kopu.** Lopta opúšťa kopačku s rýchlosťou $\vec{v}$: veľkosť $v_0$ a smer pod uhlom $\alpha$ nad zemou. Jedna šípka – ale ihrisko sa pýta na vodorovnú vzdialenosť (ako ďaleko?) a brankár na zvislú (ako vysoko?). Šípku preto rozložíme na vodorovnú zložku $v_x$ a zvislú zložku $v_y$. Nakreslíme ich spolu a vznikne pravouhlý trojuholník: $v_0$ je prepona a uhol $\alpha$ leží medzi $v_0$ a $v_x$.

<p><img style="display: block; margin-left: auto; margin-right: auto;" src='https://i.postimg.cc/Vv2KqZqP/footbal-diagram.png' width="500"></p>
<p style="text-align: center;"><em>Obr. 2.2: Rozklad kopu: rýchlosť $v_0$ pod uhlom $\alpha$ sa rozloží na vodorovnú a zvislú zložku a spolu opisujú oblúk (zdroj: vygenerované umelou inteligenciou).</em></p>


Presne na takýto trojuholník boli sínus a kosínus vymyslené. V pravouhlom trojuholníku je kosínus uhla pomerom priľahlej odvesny a prepony a sínus pomerom protiľahlej odvesny a prepony. Vynásobením každého z nich preponou $v_0$ dostaneme samotné strany:

$$v_x = v_0\cos\alpha, \qquad v_y = v_0\sin\alpha.$$

Takže $\cos\alpha$ je podiel kopu, ktorý smeruje pozdĺž ihriska, a $\sin\alpha$ podiel, ktorý smeruje nahor. Plochý kop pod uhlom 10 stupňov vloží takmer všetko do $v_x$; lopta kopnutá priamo nahor pod uhlom 90 stupňov má $\cos(90^\circ) = 0$ a celú rýchlosť zvislú; pri 45 stupňoch sú oba podiely rovnaké.

Nechajme SageMath vyčísliť zložky tohto trojuholníka pre kop rýchlosťou 15 m/s pod uhlom 30 stupňov:



````sage
# dve zložky kopu rýchlosťou 15 m/s pod uhlom 30 stupňov
N(15*cos(pi/6))
N(15*sin(pi/6))
````


> **Poznámka: uhly sa v SageMath zadávajú v radiánoch.** Uhlomer ukazuje stupne, ale `sin()`, `cos()` a ostatné goniometrické funkcie v Sage očakávajú radiány. Celý obrat má veľkosť $2\pi$ namiesto 360, takže 180 stupňov je `pi` a 30 stupňov je `pi/6`. Na ručný prevod si stačí zapamätať jednu rovnosť: $\pi\ \mathrm{rad} = 180^\circ$. Vydelením oboch strán číslom 180 dostaneme prevod stupňov na radiány, vydelením číslom $\pi$ prevod radiánov späť na stupne. SageMath vie násobiť prevodnými činiteľmi za vás, práve preto, že $\pi$ zostáva presné:
>
> ```sage
> # stupne -> radiány: násobíme pi/180
> N(30*pi/180)                     # 30 stupňov v radiánoch, 0.524
>
> # radiány -> stupne: násobíme 180/pi
> N(pi/6*180/pi)                   # pi/6 v stupňoch, 30.0
> ```
>
> Rovnaké činitele fungujú aj vo vnútri goniometrických funkcií, čo je lepšie než pamätať si hodnoty pre špeciálne uhly: `cos(45*pi/180)` je kosínus 45 stupňov, aj keď neviete, že 45 stupňov je `pi/4`.


Dostaneme 13,0 m/s pozdĺž ihriska a 7,5 m/s nahor. A keďže zložky sú odvesny trojuholníka s preponou $v_0$, Pytagorova veta ich musí zložiť späť na rýchlosť kopu:




````sage
# obe zložky sa zložia späť na pôvodnú rýchlosť
N(sqrt(15*cos(pi/6)^2 + 15*sin(pi/6)^2))   # pozor: 15*cos(...)^2 znamená (15*cos(...))^2
````



Pätnásť, teda rýchlosť, ktorou lopta opustila kopačku.

**Dva pohyby, ktoré si navzájom neprekážajú.** Rozklad sa oplatí len preto, že príroda nám vychádza v ústrety: oba smery sa navzájom neovplyvňujú. Pohyb vo vodorovnom smere nemení spôsob, akým lopta padá, a pád ju neposúva do strany. Každý smer je preto samostatná malá úloha a každú z nich ste už riešili:

- *Vodorovný smer:* ak zanedbáme vzduch, v tomto smere nepôsobí žiadna sila, takže lopta sa pohybuje stálou rýchlosťou $v_x$. Dráha je rýchlosť krát čas: $x = v_x\,t$.
- *Zvislý smer:* lopta stúpa rýchlosťou $v_y$, zatiaľ čo gravitácia ju ťahá nadol stálym zrýchlením $g$ známym zo studne v kapitole 1. Kameň pustený z pokoja tam za čas $t$ prepadol o $\tfrac{1}{2}gt^2$; tu sa ten istý pád odpočíta od stúpania.

Dosadením zložiek dostaneme celý let ako dve funkcie času:

$$x(t) = v_0\cos\alpha\, t, \qquad y(t) = v_0\sin\alpha\, t - \tfrac{1}{2}g t^2.$$

Uhol $\alpha$ rozhoduje o tom, ako sa kop rozdelí, $v_0$ o tom, aký je silný, a $g$ o tom, ako silno planéta ťahá loptu späť. Každá úloha o šikmom vrhu v tejto knihe vychádza z tejto dvojice.



## **2.3 Symbolické premenné**

Doteraz sme pod menami ukladali *čísla*. SageMath dokáže uložiť aj písmená. Príkaz `var()` zavádza symbolické premenné:




````sage
# symbolické premenné zatiaľ nemajú hodnotu; sú to písmená, s ktorými môžeme počítať
var('t alpha v0 g')
````



Tu `t` je čas, `alpha` uhol kopu, `v0` začiatočná rýchlosť a `g` gravitačné zrýchlenie. Zatiaľ ide o prázdne miesta. Každý vzorec z nich zostavený zostane vzorcom.

Teraz fyzika z oddielu 2.2 vstúpi do SageMath. Teleso vrhnuté pod uhlom $\alpha$ rýchlosťou $v_0$ sa pri zanedbaní odporu vzduchu pohybuje takto:




````sage
x(t) = v0*cos(alpha)*t
y(t) = v0*sin(alpha)*t - 1/2*g*t^2

show(x(t))
show(y(t))
````



Dve funkcie, definované presne ako pomenované čísla v kapitole 1, len teraz je každá veličina symbolická. Zátvorky v `x(t)` hovoria: funkcia závisí od `t`. Písmená `alpha`, `v0` a `g` v nej vystupujú ako parametre.



## **2.4 Kedy lopta dopadne?**

Lopta dopadne, keď jej výška klesne späť na nulu. To je rovnica a SageMath rieši rovnice symbolicky pomocou `solve()`:



````sage
solutions = solve(y(t) == 0, t)
solutions
````

````sage
# ponecháme prvé riešenie (index 0) a odrežeme jeho pravú stranu
t_land = solutions[0].rhs()
show(t_land)
````



Dve odpovede: `t == 0` (samotný kop, samozrejme platí, ale nie je užitočné) a tá zaujímavá. Riešenia sa vrátia ako zoznam, takže druhé môžeme vybrať pomocou indexu a `.rhs()` ponechá iba hodnotu: lopta je vo vzduchu $2v_0\sin\alpha / g$. Teda dvojnásobok času do najvyššieho bodu, ako sa dá z symetrie očakávať.

> **Poznámka:** `solve()` vracia zoznam *rovníc*, nie hodnôt. Indexovanie pomocou `[0]` vyberie prvú rovnicu a `.rhs()` odreže jej pravú stranu (*right-hand side*). Ak sa vám niekedy vypíše `t == ...` tam, kde ste chceli samotný výraz, zabudli ste na `.rhs()`.




## **2.5 Dostrel**

Lopta dopadne vo vodorovnej vzdialenosti $x(t_{\mathrm{land}})$. Nechajme SageMath, nech dosadí čas dopadu do funkcie letu sám:



````sage
R(alpha) = x(t_land)
show(R(alpha))
````



Funguje to, ale výsledok vyzerá nedokončene: sínusy a kosínusy sa navzájom bijú. Na to slúži `simplify()` a jeho ostrejší príbuzný `trig_reduce()`:




````sage
show(R(alpha).trig_reduce())
````



Je tu slávny výsledok aj so $\sin(2\alpha)$. Dostrel šikmo vrhnutého telesa je

$$R = \dfrac{v_0^2 \sin 2\alpha}{g}.$$

`trig_reduce()` je len jeden kľúč z celej súpravy. SageMath má celú poličku príkazov na zjednodušovanie, pretože „jednoduchý“ znamená v rôznych situáciách rôzne veci: niekedy chcete súčin, inokedy súčet, inokedy jednu goniometrickú funkciu namiesto troch. Krátka prehliadka na malých príkladoch:



````sage
var('x')

show((x^2 - 4).factor())                       # rozklad na súčin: (x + 2)*(x - 2)
show(((x + 1)^2 - x^2).expand())               # opak: roznásobenie, 2*x + 1
show((sin(x)^2 + cos(x)^2).trig_simplify())    # zlúči goniometrické funkcie: 1
show(((x - 1)/(x^2 - 1)).simplify_rational())  # skrátenie zlomkov: 1/(x + 1)
````


`factor()` rozloží polynóm na súčin a `expand()` ho znova roznásobí na súčet, `trig_simplify()` zlučuje goniometrické funkcie (opak `trig_reduce()`, ktorý ich rozkladá) a `simplify_rational()` krátí zlomky. Ak nechcete rozmýšľať, ktorý z nich sa hodí, `simplify_full()` ich zreťazí naraz a často, ale nie vždy, dospeje k čitateľnému tvaru.

> **Poznámka:** Neexistuje jediný príkaz, ktorý by vzorec urobil „jednoduchým“, pretože SageMath nemôže vedieť, čo za jednoduché považujete *vy*: `(x + 2)*(x - 2)` aj `x^2 - 4` sú správne a každý je lepší na iný účel. Postupujte tak ako my: pozrite sa na výsledok, rozhodnite sa, čo vám na ňom vadí, a vyberte nástroj práve na to.




## **2.6 Optimálny uhol**


Teraz skutočná otázka: pri ktorom $\alpha$ je $R$ najväčšie?

Nástrojom na to je *derivácia*, a preto o nej niekoľko slov skôr než sa pustíme do výpočtov, pretože príkaz, ktorý ju počíta, nehovorí, čo derivácia *je*. Derivácia krivky v bode je smernica jej dotyčnice v tomto bode: priložte k krivke rovné pravítko tak, aby sa jej dotýkalo v jedinom bode a bolo naklonené presne tak ako krivka v tomto bode; pomer zvislého prírastku k vodorovnému posunu pravítka je hodnotou derivácie. Kladná derivácia znamená, že krivka stúpa, záporná, že klesá, a čím strmšia dotyčnica, tým rýchlejšie.

Tu je spôsob, ako to *vidieť* na ľubovoľnej krivke, nielen na našom vzorci pre dostrel. Funkciu `plot()` si požičiavame z kapitoly 3 vopred; zatiaľ čítajte kód ako „nakresli tieto krivky“. Sínus je dobrý testovací prípad, pretože postupne robí všetky tri veci, ktoré krivka môže robiť:



````sage
var('x')

f(x) = sin(x)

# dotyčnica v ľubovoľnom bode x0: hodnota plus smernica krát posun
tangent(x, x0) = f(x0) + diff(f(x), x).subs(x = x0)*(x - x0)

# tri body: stúpanie, maximum, minimum
tan_rise  = tangent(x, pi/3)     # derivácia cos(pi/3) = 1/2, kladná
tan_max   = tangent(x, pi/2)      # derivácia cos(pi/2) = 0, vodorovná
tan_min   = tangent(x, 3*pi/2)    # derivácia cos(3*pi/2) = 0, vodorovná

# tri dotyčnice
sine_plot = plot(f(x), xmin=0, xmax=2*pi, color = 'blue', legend_label = '$\\sin(x)$')
tan_rise_plot = plot(tan_rise, xmin=pi/3-0.7, xmax=pi/3+0.7, color = 'red', legend_label = 'tangent at x = $\\pi/3$', linestyle='--')
tan_max_plot = plot(tan_max, xmin=pi/2-1, xmax=pi/2+1, color = 'green', legend_label = 'tangent at x = $\\pi/2$ (max)', linestyle='--')
tan_min_plot = plot(tan_min, xmin=3*pi/2-1, xmax=3*pi/2+1, color = 'purple', legend_label = 'tangent at x = $3\\pi/2$ (min)', linestyle='--')

combined_plot = sine_plot + tan_rise_plot + tan_max_plot + tan_min_plot

combined_plot.show(frame=True)
````


Všimnite si smernice. V bode $x = \pi/3$ dotyčnica stúpa spolu so sínusoidou: krivka ide nahor a sklon dotyčnice je tam hodnotou derivácie, teda kladnej. V bode $x = \pi/2$ dosahuje sínusoida vrchol a jej dotyčnica je celkom vodorovná: derivácia je nulová. V bode $x = 3\pi/2$ sínusoida dosahuje dno a dotyčnica je opäť vodorovná: aj tam je derivácia nulová. Nulová derivácia teda nevypovedá „maximum“ – vrchol aj dno ju majú. Rozdiel je v tom, čo krivka v okolí bodu *robí*:

- v **maxime** krivka stúpa, vyrovná sa a potom klesá: derivácia mení znamienko z kladného na záporné;
- v **minime** krivka klesá, vyrovná sa a potom stúpa: derivácia mení znamienko zo záporného na kladné.

To je metóda tohto oddielu na jednom obrázku: **nájdite body, v ktorých je dotyčnica vodorovná, a potom zistite, o aký druh bodu ide.**

Rovnaká hra na krivke, ktorá nás naozaj zaujíma, tentoraz s dotyčnicou pritlačenou k maximu. Dostrel kopu rýchlosťou 15 m/s na Zemi, vykreslený v závislosti od uhla:



````sage
var('a')

R_curve(a) = R(a).subs(v0 = 15, g = 9.81)

# smernica krivky dostrelu pri 30 stupňoch
slope30 = diff(R_curve(a), a).subs(a = pi/6)
tangent30(a) = R_curve(pi/6) + slope30*(a - pi/6)

# a dotyčnica v predpokladanom maxime, pri 45 stupňoch
slope45 = diff(R_curve(a), a).subs(a = pi/4)
tangent45(a) = R_curve(pi/4) + slope45*(a - pi/4)

range_plot   = plot(R_curve(a), xmin = 0, xmax = pi/2, color = 'blue', legend_label = 'range R')
tan30_plot  = plot(tangent30(a), xmin = 0, xmax = pi/2, color = 'red', legend_label = f'tangent at 30 degrees or {round(30*pi/180, 2)} radians', linestyle = '--')
tan45_plot  = plot(tangent45(a), xmin = 0, xmax = pi/2, color = 'green', legend_label = f'tangent at 45 degrees or {round(45*pi/180, 2)} radians', linestyle = '--')

combined_plot = range_plot + tan30_plot + tan45_plot

combined_plot.show(frame = True, axes_labels = ['angle [rad]', 'R [m]'])
````


Červená priamka sa dotýka modrej krivky pri 30 stupňoch a preberá jej sklon: dostrel tam ešte rastie. Zelená priamka leží vodorovne na vrchole kopca pri 45 stupňoch: derivácia je nulová presne v maxime. Posúvajte dotyčnicu rukou po krivke a pocítite, ako metóda funguje – stúpanie, vyrovnanie, prevrátenie.


SageMath vypočíta smernicu dotyčnice za nás pomocou `diff()` („differentiate“, teda „derivovať“, čiže nájsť deriváciu):



````sage
show(diff(R(alpha), alpha))
````


> **Poznámka:** Ručné derivovanie sa riadi malým súborom pravidiel: pri mocninách sa exponent presunie dopredu a zníži o jednotku, $\sin$ sa zmení na $\cos$ a súčiny či zložené funkcie majú vlastné postupy (pravidlo o derivácii súčinu a zloženej funkcie). Náš vzorec pre dostrel už vyžaduje pravidlo o derivácii zloženej funkcie, pretože obsahuje $\sin 2\alpha$. Praktická výhoda `diff()` je, že SageMath pozná všetky pravidlá a uplatňuje ich za nás: na nájdenie maxima si nemusíte pamätať ani jedno a vzorce sú niekedy aj tak príliš dlhé na ručné derivovanie.

Deriváciu položíme rovnú nule a riešime, tentoraz vzhľadom na `alpha`:




````sage
optimal = solve(diff(R(alpha), alpha) == 0, alpha)
optimal
````


Jedna množina riešení: $\alpha = \pi/4$, teda **45 stupňov** (plus celé obraty a zrkadlové kopy, ktoré dopadnú dozadu). Všimnite si, čo v odpovedi *nie je*: ani $v_0$, ani $g$. Optimálny uhol nezávisí od toho, ako silno kopnete, ani od toho, na ktorej planéte stojíte.

Nulová derivácia však hovorí iba to, že *dotyčnica je tam vodorovná*, a obrázok so sínusoidou už ukázal, čo to znamená: vrchol kopca aj dno doliny majú vodorovnú dotyčnicu. Skôr než začneme oslavovať, overme, o aký druh bodu ide. Nástrojom je *znamienko* derivácie na oboch stranách bodu, presne tá zmena znamienka, ktorú sme odčítali zo sínusoidy. Pozrime sa na smernicu krivky dostrelu pod a nad 45 stupňami, v dvoch malých krokoch namiesto jedného dlhého riadku:



````sage
# derivujeme raz a deriváciu uchováme ako samostatný vzorec
dR = diff(R(alpha), alpha)
show(dR)
````

````sage
# smernica krivky dostrelu pri 30 a pri 60 stupňoch
show(dR.subs(alpha = pi/6))    # kladná: dostrel pri 30 stupňoch ešte rastie
show(dR.subs(alpha = pi/3))    # záporná: dostrel pri 60 stupňoch už klesá
````


Pod 45 stupňami kladná, nad nimi záporná: derivácia mení znamienko z plus na mínus pri 45 stupňoch a táto zmena znamienka je znakom **maxima**.

Druhá kontrola nepotrebuje derivácie vôbec: vypočítame dostrel trochu vľavo a trochu vpravo od predpokladaného vrcholu a porovnáme (na chvíľu si požičiame čísla z oddielu 2.7):



````sage
# oba susedné uhly od 45 stupňov by mali dopadnúť kratšie ako samotných 45 stupňov
R_values = R.subs(v0 = 15, g = 9.81)
N(R_values(40*pi/180))    # plochší kop, 40 stupňov
````

````sage
N(R_values(45*pi/180))    # kandidát na maximum
````

````sage
N(R_values(50*pi/180))    # vyšší kop, 50 stupňov
````


22,6 m, 22,9 m, 22,6 m: obaja susedia prehrávajú s kandidátom uprostred.

A pre tých, ktorí uprednostňujú jeden výpočet namiesto dvoch kontrol znamienka, diferenciálny počet zhrnie kontrolu do jediného čísla: **druhej derivácie**. Ak je prvá derivácia smernicou krivky, druhá je smernica *smernice*; maximum je tam, kde smernica klesá cez nulu, takže druhá derivácia je tam záporná:



````sage
# derivujeme R dvakrát, uchováme vzorec a dosadíme 45 stupňov
d2R = diff(R(alpha), alpha, 2)
show(d2R.subs(alpha = pi/4))
````


Pre náš vzorec je to $-4v_0^2/g$, čo je záporné pre ľubovoľný kop na ľubovoľnej planéte, takže krivka je zakrivená nadol: $\pi/4$ je vrchol. A všimnite si symetriu okolo neho:




````sage
# kop pod 30 stupňami dosiahne presne tak ďaleko ako kop pod 60 stupňami
simplify(R(pi/6) - R(pi/3))
````



Nula. Dva uhly, ktorých súčet je 90 stupňov, majú rovnaký dostrel – jeden plochý a rýchly, druhý vysoký a pomalý. Osamotený zostáva len kop pod 45 stupňami.




## **2.7 Čísla pre našu futbalovú loptu**

Vzorce sú pekné, ale na futbalovom ihrisku sú metre. Hodnoty dosadíme pomocou `subs()` a aritmetiku prenecháme `N()`:



````sage
# kop rýchlosťou 15 m/s, pozemská gravitácia, 45 stupňov
N(R(pi/4).subs(v0 = 15, g = 9.81))
````



Približne `22.9` m. A keď už sme tu, vypočítajme čas letu a maximálnu výšku toho istého kopu:




````sage
# ako dlho je lopta vo vzduchu
N(t_land.subs(v0 = 15, alpha = pi/4, g = 9.81))
````

````sage
# symetria: najvyšší bod nastáva v polovici času letu
t_apex = t_land/2
N(t_apex.subs(v0 = 15, alpha = pi/4, g = 9.81))
````

````sage
# výška lopty v tomto okamihu
N(y(t_apex).subs(v0 = 15, alpha = pi/4, g = 9.81))
````


Ten istý výsledok získame aj pomocou derivácie z oddielu 2.6, len tentoraz s dosadenými číslami. Najvyšší bod je tam, kde je dotyčnica ku krivke výšky vodorovná, preto najprv zapíšme túto podmienku pre náš kop a pozrime sa na ňu skôr, než ju vyriešime:



````sage
# podmienka vodorovnej dotyčnice ku krivke výšky, s dosadenými číslami
apex_eq = diff(y(t), t).subs(v0 = 15, alpha = pi/4, g = 9.81) == 0
show(apex_eq)
````


Klesajúca priamka položená rovná nule a jej jediný koreň je najvyšší bod. Teraz ju vyriešme:



````sage
# najvyšší bod z derivácie: jedna rovnica, jeden koreň
t_apex_d = solve(apex_eq, t)[0].rhs()
N(t_apex_d)
````

````sage
# ten istý najvyšší bod, teda aj tá istá výška
N(y(t_apex_d).subs(v0 = 15, alpha = pi/4, g = 9.81))
````


Rovnakých 1,08 s a rovnakých 5,7 m: skratka cez symetriu a derivácia sa zhodujú. Skutočný brankár by poznamenal, že odpor vzduchu to všetko skracuje; vzduch vrátime do výpočtov v niektorej z neskorších kapitol.

> **Poznámka:** Všimnite si poradie krokov, ktoré sme použili, pretože je to štandardný postup pri symbolických výpočtoch vo fyzike: *sformulovať model* (dve funkcie), *vyriešiť jednu podmienku* (dopad), *dosadiť do druhej* (dostrel), *derivovať kvôli optimalizácii*, *čísla dosadiť až úplne na konci*. Čísla vstúpili práve raz, na konci. Porovnajte to s kalkulačkou, kde sa k číslam musíte zaviazať hneď v prvom riadku a každý ďalší krok dedí zaokrúhlenie.

## **2.8 Čo teraz viete**

- **Teleso vrhnuté šikmo sa rozkladá na zložky**: $v_x = v_0\cos\alpha$ pozdĺž ihriska, $v_y = v_0\sin\alpha$ nahor – odvesny pravouhlého trojuholníka s preponou $v_0$ (rýchlosť kopu); oba smery sa potom vyvíjajú nezávisle, jeden rovnomerným pohybom, druhý pádom.

- **`var()`** zavádza symbolické premenné; vzorce z nich zostavené zostávajú vzorcami a možno s nimi presne manipulovať.
- **`solve(equation, unknown)`** vráti zoznam riešení vo forme rovníc; jedno vyberiete pomocou `[i]` a pravú stranu odrežete pomocou `.rhs()`.
- **`simplify()`, `trig_reduce()`, `factor()`, `expand()`, `trig_simplify()`, `simplify_rational()`, `simplify_full()`** upravujú vzorce, každý po svojom; upravený vzorec sa ľahšie porovnáva s učebnicou.
- **`diff(expr, x)`** derivuje; derivácia je smernica dotyčnice, obrázok so sínusoidou ukázal všetky tri prípady naraz (stúpajúca dotyčnica, vodorovná v maxime, vodorovná v minime) a maximá a minimá sa nachádzajú tam, kde je derivácia nulová, čo potvrdí zmena znamienka derivácie v okolí bodu alebo znamienko druhej derivácie.
- **`.subs(name = value)`** dosadzuje čísla do vzorcov; dosadzujte raz, na konci, a nechajte `N()` dokončiť prácu.
