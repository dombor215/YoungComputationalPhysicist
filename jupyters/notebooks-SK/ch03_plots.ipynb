## **3.2 Trajektória ako krivka**

V kapitole 2 bola poloha lopty daná dvoma funkciami času: vzdialenosť $x(t) = v_0\cos\alpha \, t$, ktorá rovnomerne rastie, kým lopta letí, a výška $y(t) = v_0\sin\alpha \, t - \tfrac{1}{2}g t^2$, ktorá najprv stúpa a potom klesá. Pre obrázok chceme výšku $y$ v závislosti od vzdialenosti $x$, teda $y$ ako funkciu $x$: odložíme hodiny a ponecháme krivku. Postup je takýto: z rovnice pre vzdialenosť vyjadríme čas $t$ a dosadíme ho do výšky $y(t)$. Oba kroky sú presne to, na čo slúžilo `solve()` a dosadzovanie v kapitole 2, takže ich nerobíme ručne – znovu zostavíme obe funkcie letu a nechajme SageMath, aby $t$ eliminoval sám.

Najprv jedno rozhodnutie o pomenovaní. Rovnica, ktorú treba riešiť, je *vzdialenosť $= x$*, kde $x$ je obyčajná neznáma – lenže písmeno $x$ už obsadila funkcia vzdialenosti a `solve(x(t) == x, t)` skončí chybou. Pre SageMath nie sú funkcia a neznáma to isté, a v matematike tiež nie: $x(t)$ je pravidlo, $x$ je číslo, na ktoré sa pýtame. Trik pri takýchto kolíziách je podčiarkovník na konci: funkcie sa volajú `x_` a `y_`, neznáma zostáva `x`. V kóde ich podčiarkovník rozlišuje; `show()` ho pri sadzbe vynechá a obe sa zobrazia ako obyčajné $x$.



```sage
var('t alpha v0 g x')

# funkcie letu: podčiarkovník v kóde, v show() obyčajné x a y
x_(t) = v0*cos(alpha)*t
y_(t) = v0*sin(alpha)*t - 1/2*g*t^2

# vyriešime "vzdialenosť = x" vzhľadom na čas, ponecháme (jediné) riešenie
t_hit = solve(x_(t) == x, t)[0].rhs()
show(t_hit)
```

```sage
# dosadíme čas do výšky: y ako funkcia x
y_traj(x) = y_(t).subs(t = t_hit).trig_simplify()
show(y_traj(x))
```



Parabola, ako to sľubuje každý učiteľ fyziky. Najjednoduchší graf v SageMath sa kreslí príkazom `plot()`: jeho argumentmi sú to, čo kresliť, a rozsah vodorovnej osi, zapísaný ako `xmin` a `xmax`. Mená hovoria, čo robia – rovnakým štýlom ako `ymin` a `ymax`, ktoré uvidíme nižšie.




```sage
# jeden kop pod uhlom 45 stupňov rýchlosťou 15 m/s: najprv dosadíme čísla
kappa45(x) = y_traj(x).subs(alpha = pi/4, v0 = 15, g = 9.81)
plot(kappa45(x), xmin = 0, xmax = 25)
```



Osi sa škálujú automaticky a krivka tam je: hore, cez vrchol, dole.

> **Poznámka:** `plot()` musí vedieť, *kde funkciu vyhodnocovať*, takže interval nie je voliteľný. Ak sa SageMath sťažuje, že nevie, čo kresliť, takmer vždy to znamená, že ste zabudli na `xmin = ...` alebo `xmax = ...`.




## **3.3 Všetky uhly naraz**

Jedna krivka je jeden uhol. Brankár chce celý súbor: uhly od 20 do 70 stupňov. Nástrojom na opakovanie výpočtu s rôznymi hodnotami je list comprehension (zostavenie zoznamu), `[expression for item in list]`, s ktorým sa tu stretávame prvýkrát. Funguje aj pre grafy, pretože v SageMath je graf objekt ako každé iné číslo:



```sage
angles = [20, 30, 40, 50, 60, 70]      # v stupňoch

# jeden objekt grafu pre každý uhol, zozbieraný do zoznamu
plots = [plot(y_traj(x).subs(alpha = a*pi/180, v0 = 15, g = 9.81),
              xmin = 0, xmax = 25,
              color = hue(a/90),
              legend_label = str(a) + ' degrees')
         for a in angles]
```



Každý graf dostal `color` a `legend_label`, aby sme ich vedeli rozlíšiť. Grafy sa spájajú jednoduchým sčítaním a súčet sa zobrazí pomocou `show()`. Zvyk, ktorý stojí za to si osvojiť hneď tu: *súčet zostavte, pomenujte ho a až potom zobrazte*. Súčet je objekt grafu ako každý iný, a preto si zaslúži meno ako každý iný; nasledujúce oddiely budú k tomuto obrázku ďalej pridávať a nikto nechce kvôli tomu znovu písať list comprehension.




```sage
# operátor + prekrýva grafy; súčet uložíme pod menom
family_plot = sum(plots)

show(family_plot, ymin=0)
```



Šesť parabol s jasnou zákonitosťou: 20 a 70 stupňov dopadnú spolu, 30 a 60 spolu a 40 s 50 takmer spolu, presne podľa symetrie $\sin 2\alpha$ z kapitoly 2. Najvyšší oblúk nedopadne ďalej ako najplochejší.




## **3.4 Obálka: kam lopta nedopadne**

Pozrite sa na súbor kriviek a uvidíte strop: vonkajšiu hranicu, ktorej sa lopty dotýkajú, ale nikdy ju nepretnú. Nazýva sa *obálka* a SageMath ju vie nielen nakresliť, ale aj vypočítať. Trik spočíva v substitúcii, vďaka ktorej je algebra príjemnejšia. Zapíšeme $u = \tan\alpha$ a použijeme vzťah $1/\cos^2\alpha = 1 + \tan^2\alpha$:



```sage
var('u')

# tá istá trajektória, ale vyjadrená pomocou u = tan(alpha)
y_u(x) = u*x - g*x^2/(2*v0^2)*(1 + u^2)
show(y_u(x))
```



Pre pevne zvolenú vzdialenosť $x$ je najvyšší kop ten s najlepším $u$. Maximum funkcie $y_u$ vzhľadom na $u$ nastáva tam, kde je jej derivácia podľa $u$ nulová:




```sage
u_best = solve(diff(y_u(x), u) == 0, u)[0].rhs()
show(u_best)
```



Dosadením tohto optimálneho kopu späť do trajektórie dostaneme obálku:




```sage
envelope(x) = y_u(x).subs(u = u_best)
show(envelope(x))
```



$$y_{env} = \dfrac{v_0^2}{2g} - \dfrac{g\,x^2}{2v_0^2}$$

Opäť parabola, ale otočená dolu hlavou, ktorá leží nad všetkými ostatnými. Každý jej bod patrí jednej lopte, tej, ktorá bola kopnutá pod práve takým uhlom, aby sa tam dostala.

Teraz nakreslime všetko spolu. Grafy majú možnosť `linestyle`; `'--'` znamená prerušovanú čiaru:




```sage
env_plot = plot(envelope(x).subs(v0 = 15, g = 9.81), xmin = 0, xmax = 25,
                color = 'black', linestyle = '--', legend_label = 'envelope')

# predchádzajúci súbor kriviek plus obálka: jeden pomenovaný obrázok
combined_plot = family_plot + env_plot

show(combined_plot, figsize = 5, ymin = 0)
```



Tu je brankárov obrázok: všetko vo vnútri prerušovanej paraboly možno zasiahnuť pod nejakým uhlom, všetko nad ňou je v bezpečí. Vrchol obálky, $v_0^2/2g$ pri $x = 0$, je najvyšší bod, ktorého môže lopta rýchlosťou 15 m/s vôbec dosiahnuť – pri kope priamo nahor.




## **3.5 Ako graf urobiť čitateľným**

Graf, ktorý sa nedá prečítať, je zbytočný výpočet. Tu sa zhodnotí zvyk pomenúvať: ten istý `combined_plot` z predchádzajúceho oddielu, lepšie upravený, bez opätovného zostavovania čo i len jednej krivky. Možnosti sú tie každodenné:



```sage
show(combined_plot,
     figsize = 6,             # veľkosť obrázka
     ymin = 0, ymax = 12,     # rozsah zvislej osi
     frame = True,            # rámik okolo obrázka namiesto osí
     axes_labels = ['$x$ [m]', '$y$ [m]'],   # popisy osí
     gridlines = True)        # mriežka na pozadí
```



Tri detaily stoja za to, aby sa z nich stal zvyk. Po prvé, `axes_labels` akceptujú LaTeX: popis napíšte medzi znaky dolára a `$x$ [m]` na osi vyjde vysádzané matematickým písmom, kurzívou, ktorú poznáte zo vzorcov. Po druhé, `axes_labels` je miesto, kde patria jednotky, podľa našej dohody z kapitoly 1: čísla v notebooku zostávajú holé, jednotky žijú v popisoch. Po tretie, `ymin = 0` nie je kozmetika: bez neho SageMath pokojne zobrazí loptu dopadajúcu pod úroveň zeme.

> **Poznámka:** Farby možno zadať ako `hue(number)` s číslom medzi 0 a 1, čo rozmiestni súbor kriviek rovnomerne po dúhe, alebo názvami ako `'red'`, `'black'`. `legend_label` je obyčajný text, takže ak má obsahovať číslo, zostavuje sa pomocou `str(a) + ' degrees'`.

Krivky hovoria „niekde tam hore“; bodka hovorí „presne tu“. Stojí za to vyznačiť *vrcholy* kopov, teda body, v ktorých lopta prestáva stúpať. Ich čas poznáme z kapitoly 2: polovica času letu, $t_{\mathrm{apex}} = v_0\sin\alpha / g$. Pre obrázok potrebujeme *miesto* každého vrcholu, dvojicu (vzdialenosť, výška). Najčistejším nástrojom na to je `lambda`. Je súčasťou Pythonu, nie matematiky SageMath: pomocou nej možno jednoducho zapísať krátke postupy, ktoré prijmú jeden alebo viac vstupov, spracujú ich a vrátia výstup v ľubovoľnej podobe – tu jeden vstup (uhol v stupňoch) a jedno číslo na výstupe, pričom náš kop (`v0 = 15`, `g = 9.81`) a prevod stupňov na radiány sú zabudované vnútri:



```sage
# čas vrcholu, zo symetrie v kapitole 2
t_apex = v0*sin(alpha)/g

# krátky postup pre každú súradnicu: na vstupe uhol v stupňoch, na výstupe miesto v metroch
x_apex = lambda a: N(x_(t_apex).subs(alpha = a*pi/180, v0 = 15, g = 9.81))
y_apex = lambda a: N(y_(t_apex).subs(alpha = a*pi/180, v0 = 15, g = 9.81))

show(x_apex(45))
show(y_apex(45))
```



Definíciu čítajte zľava: kľúčové slovo `lambda`, meno vstupu (`a`, uhol v stupňoch), dvojbodka a to, čo sa má vypočítať. Lambda *nie je* matematická funkcia ako `x_(t) = ...`: neobsahuje žiadnu rovnicu na riešenie, nič na derivovanie, žiadny vzorec na zobrazenie – je to čistý výpočet. Kop pod uhlom 45 stupňov má vrchol 11,5 m od miesta kopu a 5,7 m vysoko.

> **Poznámka:** Ak už trochu poznáte Python, možno ste sa stretli s príkazom `def`: `def f(a): return ...` funguje rovnako, s telom zapísaným na jednom alebo viacerých odsadených riadkoch. `lambda` je tá istá myšlienka stlačená do jedného výrazu – správny nástroj, keď sa postup zmestí na jeden riadok, a `def`, keď sa nezmestí.

Odmena prichádza v zozname nižšie: keďže všetko dosadzovanie je ukryté vo vnútri postupov, zostáva zoznam tak plochý ako zoznam parabol v oddiele 3.3. A tentoraz posunieme uhly ďalej, až do 90 stupňov:



```sage
# jedna dvojica (vzdialenosť, výška) pre každý uhol
# uhly z predchádzajúceho plus kop priamo nahor
apex_angles = angles + [80, 90]

apexes = [(x_apex(a), y_apex(a)) for a in apex_angles]

show(apexes)
```



Osem dvojíc, od `(7.4, 1.3)` pri 20 stupňoch až po `(0, 11.5)` pri 90 stupňoch. Kop pod 90 stupňami nemá v predchádzajúcom súbore žiadnu parabolu: lopta kopnutá priamo nahor nezíska žiadnu vzdialenosť a požiadavka na `y_traj` pre tento uhol by znamenala delenie hodnotou $\cos 90^\circ = 0$. Jeho vrchol je napriek tomu úplne dobrý bod – nulová vzdialenosť, plná výška. Takýto zoznam sa kreslí pomocou `point()`, ktorý chce presne tento tvar – zoznam dvojíc $(x, y)$ – a nič iné:




```sage
# jedna bodka pre každý vrchol, nad obrázkom
apex_plot = point(apexes, size = 40, color = 'red')

show(combined_plot + apex_plot, figsize = 5, ymin = 0)
```



Každá bodka leží na svojej vlastnej parabole a bodka pre 90 stupňov leží presne na vrchole obálky, $v_0^2/2g$: najvyšší bod obrázka patrí lopte kopnutej priamo nahor.

Bodky označujú miesta; čiara rozpráva príbeh pozdĺž nich. `list_plot()` prijíma ten istý zoznam dvojíc a pridáva jeden parameter, ktorý stojí za to poznať: `plotjoined = True` vedie čiaru cez bodky:



```sage
# tie isté dvojice spojené do jednej čiary
apex_line = list_plot(apexes, plotjoined = True, color = 'red')

show(combined_plot + apex_plot + apex_line, figsize = 5, ymin = 0)
```



Čiara robí zákonitosť viditeľnou: vrcholy stúpajú po hladkom oblúku, ktorý končí na vrchole obálky. Tento oblúk je polovica elipsy – vrchol každého uhla kopu na ňom leží a jeho najvyšší bod je samotný strop.

> **Poznámka:** `point()` na hŕstku vyznačených bodov; `list_plot()` s `plotjoined = True`, keď body patria k jednému príbehu.

## **3.6 Čo teraz viete**

- **`plot(expr, xmin, xmax)`** kreslí funkciu jednej premennej; rozsah osi je povinný a jeho mená zodpovedajú `ymin` a `ymax`.
- **Podčiarkovník na konci** (`x_`, `y_`) umožňuje funkcii a symbolu zdieľať písmeno; `show()` oba vysádza ako obyčajné $x$.
- **Objekty grafov možno ukladať do zoznamov** (list comprehension) a **sčítať** pomocou `+`, čím sa prekryjú súbory kriviek; súčet pomenujte (`family_plot`, `combined_plot`) skôr než ho zobrazíte pomocou `show()`, aby sa dal znovu použiť a rozširovať.
- **`point(pairs)`** vyznačí zoznam bodov $(x, y)$ na obrázku, tu vrcholy trajektórií; **`list_plot(pairs, plotjoined = True)`** spojí bodky do čiary.
- **`lambda a: ...`** je jednoriadkový postup z Pythonu – na vstupe hodnoty, na výstupe spracovaný výsledok v ľubovoľnej podobe; nie je to matematická funkcia, takže nemá čo riešiť ani zobrazovať. Príkaz `def` funguje rovnako pre viacriadkové postupy.
- **`color`, `linestyle`, `legend_label`, `axes_labels`, `figsize`, `gridlines`, `ymin`, `frame`** sú každodenné možnosti formátovania; `axes_labels` akceptujú LaTeX medzi znakmi dolára.
- **Obálku súboru kriviek** nájdeme maximalizáciou podľa parametra súboru, tu pomocou substitúcie a `diff()` + `solve()`: symbolický výsledok, nakreslený až na konci.
- **Jednotky patria do popisov osí**, nie do samotných čísel, podľa dohody z kapitoly 1.
