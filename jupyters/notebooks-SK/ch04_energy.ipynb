## **4.2 Energia v dvoch vzorcoch**

Vozík s hmotnosťou $m$ vo výške $y$, pohybujúci sa rýchlosťou $v$, má



```sage
var('m g y v y0')

E_pot = m*g*y                    # potenciálna energia
E_kin = 1/2*m*v^2                # kinetická energia
E_initial = m*g*y0               # celková energia: vozík v pokoji vo výške y0
```



Zákon zachovania energie hovorí, že ich súčet sa nikdy nemení, pretože na vozík na tejto horskej dráhe nekoná prácu nič: ani trenie, ani motor. Na začiatku vozík stojí vo výške $y_0$, takže celková energia je $E = mgy_0$, čo je číslo, ktoré bunka vyššie ukladá pod menom `E_initial`. V ľubovoľnom inom mieste trate:




```sage
# rýchlosť ako funkcia výšky vyplýva zo zákona zachovania energie
v_of_y_all = solve(E_kin + E_pot == E_initial, v)
show(v_of_y_all)
```

```sage
# kladný koreň
v_of_y = v_of_y_all[1].rhs()
show(v_of_y)
```


Kladný koreň: $v = \sqrt{2g\,(y_0-y)}$. Všimnite si, čo vypadlo: hmotnosť $m$ je preč. Rýchlosť nezávisí od toho, aký je vozík ťažký, rovnako ako kladivo a pierko padajúce vo vákuu. A všimnite si, čo zostalo symbolom: začiatočná výška $y_0$. Vzorec platí pre ľubovoľný kopec; o ktorý konkrétny kopec ide, rozhodujú hodnoty, a tie zapíšeme do jedného zoznamu rovníc:



```sage
# číselné hodnoty parametrov zhromaždené v jednom zozname
values = [y0 == 20,               # m, začiatočná výška
          g == 9.81]              # m/s^2, normálne tiažové zrýchlenie
```


> **Poznámka:** Zoznam `values` je trik, ktorý stojí za to prevziať. V Sage je rovnica objekt ako každý iný, vrátane `y0 == 20`, takže rovnice možno ukladať do zoznamu – a `.subs()` bez problémov spracuje celý zoznam naraz. Robí to isté ako zápis `.subs()` s kľúčovými slovami, ale čísla teraz žijú na jednom pomenovanom mieste: keď pribudne nový parameter, pripojí sa do zoznamu, a keď kopec zväčšíme, zmení sa jeden riadok a každé neskoršie `.subs(values)` ho nasleduje. Každá rovnica má v komentári svoju jednotku, takže zoznam slúži zároveň ako malá tabuľka fyzikálnych vstupov.

Hneď dve čísla:




```sage
# rýchlosť na dne údolia (y = 0)
speed_bottom = v_of_y.subs(y = 0).subs(values)

# v m/s
speed_bottom
```

```sage
# v km/h
speed_bottom*3.6
```

```sage
# rýchlosť na vrchole 10 m vysokej vyvýšeniny
speed_top = v_of_y.subs(y = 10).subs(values)

# v m/s
speed_top
```

```sage
# v km/h
speed_top*3.6
```


Takmer 70 km/h na dne a asi 50 km/h cez vyvýšeninu. Všimnite si aj tvar výsledku: rýchlosť závisí iba od *výšky*, o ktorú vozík klesol, nie od toho, či bol zostup strmý alebo mierny. Dno údolia by mohlo byť naklonené, zakrivené, akékoľvek; vo rovnakých výškach sú rýchlosti rovnaké.




## **4.3 Tri druhy energie v jednom grafe**

Aby sme videli celý príbeh naraz, nakreslime potenciálnu, kinetickú a celkovú energiu pozdĺž trate. Samotnú trať opíšeme výškovým profilom $y(x)$, čiže ľubovoľnou hladkou vlnitou funkciou so štartom vo výške 20 m, údolím a 10 m vysokou vyvýšeninou. Tu je jedna, zostavená z kosínusov:



```sage
var('x')

# vymyslená trať: začína vo výške 20 m, klesá na 5 m, vyvýšenina vo výške 10 m
y_track(x) = piecewise([((0, 20), 7.5*cos(2*pi*x/40) + 12.5),
                        ((20, 80), 2.5*cos(2*pi*x/40) + 7.5)])

p_track = plot(y_track(x), xmin=0, xmax=80, ymin=0, ymax=25, axes_labels = ['$h\\;\\mathrm{[m]}$', '$x\\;\\mathrm{[m]}$'], 
               color='green', frame=True, gridlines=True)
p_track
```


Novým nástrojom je tu `piecewise`. Prijíma zoznam dvojíc, pričom každá dvojica obsahuje interval a vzorec, ktorý na ňom platí. Keď sa spýtame na hodnotu, zistí, do ktorého intervalu vstup patrí, a vyhodnotí príslušný vzorec. Naša trať presne toto potrebuje: jeden kosínus, ktorý privedie vozík do údolia (0 až 20 m), a plochší kosínus nad vyvýšeninou (20 až 80 m) – dva úseky zošité do jednej po častiach definovanej funkcie, pre každý interval zvlášť. V správnom matematickom zápise by po častiach definovaná funkcia vyzerala takto:

$$
y_{\mathrm{track}}(x) =\left\{ \begin{array}{ll}
7{,}5\cos\left(\dfrac{2\pi x}{40}\right) + 12{,}5, & 0 \le x \le 20 \\
2{,}5\cos\left(\dfrac{2\pi x}{40}\right) + 7{,}5, & 20 < x \le 80
\end{array} \right.
$$


Samotná výška, ako obrázok. Teraz energie; výškový profil je všetko, čo na ne potrebujeme:




```sage
E_track(x) = E_pot.subs(y = y_track(x))          # potenciálna energia pozdĺž trate
E_k_track(x) = E_initial - E_track(x)            # kinetická = celková - potenciálna
```



Kinetická energia ako „celková mínus potenciálna“ je druhé použitie zákona zachovania energie a stojí za to zastaviť sa pri ňom. Súčet $E = E_k + E_p$ je rovnaké číslo v každom bode trate, $E = mgy_0$ zo štartu v pokoji. Takže keď poznáme potenciálnu energiu $E_p(x) = mg\,y(x)$ v polohe $x$, kinetická energia je jednoducho to, čo zostane z celkového rozpočtu: $E_k(x) = E - E_p(x)$. Riadok kódu `E_k_track(x) = E_initial - E_track(x)` je táto jediná veta zapísaná v Sage.

Všimnite si, čoho sa nikdy nedotkneme: rýchlosti. Na zakrivenej trati by bolo priame vyčíslovanie $v$ v každom bode $x$ naozaj nepríjemné. Rýchlosť $\vec{v}$ je vektor: ako vozík ide, jej smer sa otáča so sklonom trate a obe zložky sa plynule menia, $v_x$ aj $v_y$ po svojom. Veľkosť rýchlosti by sa musela dodatočne poskladať zo zložiek, $v = \sqrt{v_x^2 + v_y^2}$, čo si najprv vyžaduje obe zložky ako funkcie $x$, a takú algebru nikto nechce. Energie sa tomu vyhnú, pretože sú skalárne veličiny: $E_p = mgy$ a $E_k = \tfrac{1}{2}mv^2$ sú jednotlivé čísla bez smeru a sčítajú sa a odčítajú ako čísla. V tom spočíva tichá sila energetickej metódy – mení vektorovú úlohu na skalárne odčítanie. Teraz všetky tri v jednom obrázku. Hmotnosť sa v oddiele 4.2 vykrátila, takže nepotrebovala číselnú hodnotu; teraz ju potrebuje a pripojí sa do zoznamu:




```sage
# hmotnosť vozíka sa pripojí do zoznamu parametrov
values += [m == 500]              # kg

p1 = plot(E_track(x).subs(values), xmin = 0, xmax = 80, color = 'blue',
          legend_label = 'potential energy')
p2 = plot(E_k_track(x).subs(values), xmin = 0, xmax = 80, color = 'red',
          legend_label = 'kinetic energy')
p3 = plot(E_initial.subs(values), xmin = 0, xmax = 80, color = 'black', linestyle = '--',
          legend_label = 'total energy')

p_format = plot([], '$x\\;\\mathrm{[m]}$', '$E\\;\\mathrm{[J]}$', ymax=120e3, frame=True, gridlines=True)

p_energies = p1 + p2 + p3 + p_format

show(p_energies)
```


> **Poznámka:** Riadok s `p_format` je malý trik. `plot([])` nakreslí prázdny zoznam kriviek, takže nekreslí nič, ale stále akceptuje všetky možnosti formátovania ako skutočný graf. Keď sa grafy spájajú pomocou `+`, možnosti všetkých sčítancov sa zlúčia, takže prázdny graf prenesie rámik, mriežku, popisy osí a `ymax` do výsledného obrázka. Tie isté možnosti by mohli byť aj v ktoromkoľvek z grafov `p1`, `p2` alebo `p3` a výsledok by vyzeral rovnako; uchovať ich v samostatnom prázdnom grafe je len trochu flexibilnejšie, keď sa kombinácia grafov mení. Vynechajte jednu energiu, pridajte štvrtú krivku, zobrazte samostatne `p1` a `p2` – formátovanie putuje ďalej s `p_format` a nie je treba ho prepisovať do toho grafu, ktorý práve zostal.

> **Poznámka:** Jeden riadok, `values += [m == 500]`, a hmotnosť je súčasťou príbehu. Sčítanie dvoch zoznamov pomocou `+` ich zlepí do jedného dlhšieho zoznamu a `+=` rozšíri `values` na mieste, takže každé neskoršie `.subs(values)` vidí aj hmotnosť. Chcete ponechať starý zoznam nedotknutý? `values_updated = values + [m == 500]` vytvorí nový zoznam a `values` ostane tak, ako bol. Čitatelia, ktorí poznajú Python, poznajú aj `.append()`: tiež pridá jediný prvok na koniec, na mieste – `values += [...]` robí tú istú prácu obyčajným sčítaním zoznamov, s bonusom, že môže prijať viacero rovníc naraz, `values += [m == 500, r == 15]`.



```sage
# graf trate a energií nad sebou pomocou graphics array
# parameter figsize určuje rozmery výsledného obrázka
graphics_array([[p_track], [p_energies]]).show(figsize=[10,7])
```



Prečítajte si obrázok: modrá vlna potenciálnej energie a červená vlna kinetickej energie sú zrkadlovými obrazmi, vymieňajú si výšku za rýchlosť a naopak, zatiaľ čo prerušovaná čierna čiara celkovej energie je dokonale vodorovná. Táto vodorovná čiara *je* zákon zachovania energie, nakreslený. Tam, kde trať klesá do údolia, je modrá nízko a červená vysoko: vozík je nízko a rýchly. Nad vyvýšeninou sa stretnú v polovici.

*Obr. 4.2: Tri druhy energie, jeden zákon: červená kinetická a modrá potenciálna vlna sa vymieňajú presne, zatiaľ čo prerušovaná čiara celkovej energie je vodorovná.*

> **Poznámka:** Kinetická energia nikdy nesmie byť záporná, $E_k = E - mg\,y(x) \ge 0$, takže vozík nemôže nikdy vystúpiť nad svoju začiatočnú výšku $y = 20~\mathrm{m}$. Keby niekto do tejto trate zabudoval 25 m vysoký kopec, červená krivka by klesla pod nulu a obrázok by klamal. O tom, čo je fyzikálne, rozhoduje rovnica, nie kresba.




## **4.4 Odlet z konca trate**

Teraz trať náhle končí na vrchole 10 m vysokej vyvýšeniny a vozík sa pohybuje vodorovne rýchlosťou, ktorú sme tam vypočítali. Z kapitoly 2 vieme o vodorovnom vrhu všetko: je to šikmý vrh s nulovým uhlom. Funkcie letu so začiatočnou rýchlosťou získanou z energie:



```sage
# vodorovný vrh z výšky 10 m rýchlosťou na vrchole vyvýšeniny
v_launch = N(v_of_y.subs(y = 10).subs(values))
v_launch
```

```sage
var('t')

x_flight(t) = 80 + v_launch*t
y_flight(t) = 10 - 1/2*g*t^2

# dopad: y_flight == 0, vyriešené symbolicky
t_land = solve(y_flight(t) == 0, t)[1].rhs()
N(t_land.subs(values))
```

```sage
# kde dopadne
N(x_flight(t_land).subs(values))
```



Asi 1,43 s vo vzduchu, dopad zhruba 20 m za koncom trate. A rýchlosť dopadu? Opäť energia, bez potreby zložiek:




```sage
# pri dopade y = 0: celý začiatočný rozpočet je kinetický
N(solve(1/2*m*v^2 == E_initial, v)[1].rhs().subs(values))
```



Tých istých 19,8 m/s ako na dne údolia: čo ide dolu, vráti sa s rovnakou rýchlosťou, výška za výšku. Účtovanie energie zredukuje aj dopad na jediný riadok.
