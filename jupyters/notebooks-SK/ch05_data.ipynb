## **5.1 Zadanie**

> Pružina sa ručne naťahuje na dĺžky $1, 2, \ldots, 10~\mathrm{cm}$ a pri každej dĺžke sa silomerom desaťkrát odčíta ťahová sila. Vznikne tak $10\times10=100$ údajov, dvojíc $(x, F)$. Rastie sila lineárne s predĺžením a aká tuhosť pružiny $k$ sa ukrýva v číslach?

Fyzika je na jeden riadok, Hookov zákon: sila $F$ potrebná na natiahnutie pružiny o $x$ je

$$F = kx,$$

kde $F$ je sila v $\mathrm{N}$, $k$ tuhosť pružiny v $\mathrm{N/m}$ a $x$ predĺženie v $\mathrm{m}$. Samotné meranie je poctivé, ale hrubé: ruka ťahá, silomer sa chvejí, oko zaokrúhľuje a každý zo sto údajov nesie svoju dávku neistoty. Tá neistota tu nie je na obtiaž, je to materiál. Táto kapitola ju zadá do počítača, skontroluje, vyčistí, nakreslí a nakoniec spriemeruje.

A $k$ stojí za to poznať. Tlmič auta je pružina v prestrojení a jej tuhosť rozhoduje, či sa výmoľ dostane k cestujúcim ako úder alebo len ako pokrčenie ramien. Príliš tuhá pružina trestá, príliš mäkká nedovolí kolesu sledovať cestu. Niekde v sto údajoch nižšie sa skrýva číslo, ktoré by inžinier potreboval.

Po načítaní sú údaje `DataFrame`, pandasovská verzia tabuľky z Excelu:

V dátovej vede sa tento krok nazýva *predspracovanie dát* (*data preprocessing*) alebo *tvorba príznakov* (*feature engineering*): pretvorenie surovej tabuľky tak, aby ju akákoľvek analýza na ďalšom stupni našla pripravenú. Prevod jednotiek by sa dal urobiť aj neskôr ručne, na každej hodnote, ale potom by si každé použitie dĺžky muselo pamätať prevodný činiteľ. Nový stĺpec urobí prácu raz, tam, kde dáta žijú. `sample(n=10)` zobrazí desať náhodných riadkov, pri každom spustení iných. Tabuľka ostáva, mení sa len pohľad.

## **5.7 Spriemerovanie opakovaní**

Desať údajov pri jednej dĺžke je desať pohľadov na to isté ťahanie. Rozptyl medzi nimi je ruka a silomer, nie pružina. Aby sme videli pružinu, údaje každej dĺžky sa zhustia do priemeru a rozptylu:



```sage
# štatistiky z dát – priemery a smerodajná odchýlka (std)
stats = data.groupby("x [m]")["F [N]"].agg(["mean", "std"]).reset_index()
stats
```


Riadok robí štyri úlohy, jednu po druhej:

1. `data.groupby("x [m]")` rozdelí sto riadkov do desiatich skupín, jednej na dĺžku.
2. `["F [N]"]` vyberie stĺpec, nad ktorým sa počíta.
3. `.agg(["mean", "std"])` žiada dve čísla na skupinu: priemer a smerodajnú odchýlku jej síl.
4. `.reset_index()` zmení výsledok späť na obyčajnú tabuľku, v ktorej je `x [m]` bežným stĺpcom.

Výsledkom je nový `DataFrame`, `stats`, jeden riadok na dĺžku: kde sa údaje danej dĺžky v priemere nachádzajú a ako ďaleko sa typický údaj od tohto priemeru odchyľuje, v newtonoch. To isté sa dá urobiť aj v Exceli. Urobiť to jedným riadkom, ktorý prežije zmenu dát, je dôvod, prečo táto kapitola existuje.



## **5.8 Chybové úsečky vlastnej výroby**

Priemer bez rozptylu je pol čísla. Rozptyl patrí do grafu ako *chybová úsečka* (*error bar*): malá čiara cez bod, ktorá ide od hodnoty mínus jej neistota po hodnotu plus jej neistota. Sage na to nemá hotový príkaz, čo robí z chybovej úsečky dobrú zámienku zoznámiť sa s vlastnou funkciou: receptom s menom, parametrami na vstupe a jednou hodnotou na výstupe. Jedna funkcia na jednu os:



```sage
# vodorovná úsečka: x ide od x - dx do x + dx, vo výške y
def x_errorbar(x, y, dx):
    return line([(x - dx, y), (x + dx, y)], color='lime', thickness=1.5)

# zvislá úsečka: y ide od y - dy do y + dy, v polohe x
def y_errorbar(x, y, dy):
    return line([(x, y - dy), (x, y + dy)], color='lime', thickness=1.5)
```


`def` dá receptu meno, zátvorky vymenúvajú, čo recept potrebuje vedieť, a odsadené telo sa vykoná, až keď sa funkcia zavolá. Volanie ako `y_errorbar(0.06, 15.2, 1.2)` vypočíta jednu čiaru a vráti ju. To je práca príkazu `return`: odpoveď funkcie je hodnota ako každá iná, pripravená na uloženie, zaradenie do zoznamu alebo sčítanie. Tri čísla vo volaní vyplnia tri parametre v poradí: poloha, výška, polovičná dĺžka úsečky.

Neistoty sú vlastné tejto kapitole. Dĺžky sa značili metrovou latou s presnosťou na pol milimetra: $\Delta x = 0{,}0005~\mathrm{m}$. Rozptyl sily je už vypočítaný, je v stĺpci `std` tabuľky `stats`. Funkciám dáme desať priemerov, jedno zostavenie zoznamu (*list comprehension*) na os:



```sage
# chybové úsečky na priemeroch: x s presnosťou na pol milimetra, y na jednu smerodajnú odchýlku
x_bars = [x_errorbar(xm, m, 0.0005) for xm, m in zip(stats['x [m]'], stats['mean'])]
y_bars = [y_errorbar(xm, m, s) for xm, m, s in zip(stats['x [m]'], stats['mean'], stats['std'])]

# priemery ako veľké červené body nad surovými dátami
plot_means = points(list(zip(stats['x [m]'], stats['mean'])), size=40, color='red', zorder=-200)

err_plot = plot_means + measurement_plot + sum(x_bars + y_bars)
err_plot.show(figsize=12)
```


Volanie `.show()` sa končí parametrom `figsize` a toto slovo je nové, takže si zaslúži pozornosť. `figsize` nastavuje fyzickú veľkosť obrázka v palcoch: dvojica ako `figsize=[12, 6]` znamená dvanásť palcov na šírku a šesť na výšku. Jediné číslo, ako tu, znamená štvorec, `figsize=12` stojí za `[12, 12]`. Samotná scéna sa neroztiahne: SageMath zachováva pomer strán grafu a kreslí rámik vo vnútri obrázka, takže dáta sa nikdy nedeformujú, rastie len plátno.

Plátno tu rásť musí, pretože obrázok je hustý: sto sivých bodov, desať červených priemerov a dvadsať tenkých úsečiek zdieľa jeden rámik. Na malom predvolenom plátne by sa vodorovné úsečky, dlhé sotva milimeter, stratili. Zväčšenie obrázka zväčšuje s ním každý popis, značku aj úsečku, takže pri `figsize=12` sú chybové úsečky jasne viditeľné. `figsize` funguje rovnako všade, kde sa graf objaví: v `show()`, v `save()` alebo priamo v `plot()` a `points()`.

> **Poznámka:** Časti sa spájajú sčítaním: v SageMath jedna grafika plus druhá nakreslí obe do toho istého obrázka. `sum(bars, start)` je obyčajný Python, ktorý toto sčítanie robí v cykle, počnúc hodnotou `start`, tu surovým rozptylom s červenými priemermi, a pripája k nemu každú úsečku. Funkcie sú teraz majetkom kapitoly: ktorýkoľvek graf, ktorý potrebuje pripojiť neistotu, ich môže zavolať.

> **Poznámka:** Zvislá úsečka tu nakreslená je jedna smerodajná odchýlka *údajov*. *Priemer* z desiatich je stabilnejší než ktorýkoľvek jednotlivý údaj, jeho vlastná neistota je $\mathrm{std}/\sqrt{10}$, štandardná chyba priemeru. Cvičenie 5.4 sa pýta, čo obrázok hovorí, keď sa nakreslí táto úsečka.



## **5.9 Preloženie modelu**

Teraz k samotnej tuhosti pružiny. Model je $F = kx$ s neznámym $k$ a v Sage sa model napíše skôr, než sa použije:



```sage
# premenné vystupujúce v modeli – tuhosť pružiny k a predĺženie x
k = var("k")
x = var("x")

# definovanie modelu
F(x) = k*x

# zobrazenie na kontrolu
show(F)
```


Čo vstupuje do preloženia? Môže to byť všetkých sto surových bodov z `data_points`. Pokojnejšou voľbou je desať priemerov, jeden na dĺžku: opakovania sú už spriemerované a preloženie kladie desať otázok namiesto sto. Preložené body teda sú:



```sage
# preložené dáta: jedna dvojica (x, priemer F) na dĺžku
x_F_means = list(zip(stats['x [m]'], stats['mean']))
x_F_means
```


Samotné preloženie je jeden príkaz. `find_fit()` berie ako prvý zoznam bodov a ako druhý model a hľadá hodnoty parametrov, ktoré model čo najviac priblížia k bodom:



```sage
# nájdenie parametra k, výstup vo forme slovníka Pythonu
fit = find_fit(x_F_means, F, solution_dict=True)
fit
```


> **Poznámka:** S `solution_dict=True` sa odpoveď vráti ako slovník Pythonu, dvojica `kľúč: hodnota`, ktorú neskorší kód číta podľa mena, ako v `fit[k]`. Ako vyzerá výstup bez neho alebo so `solution_dict=False`? Vyskúšajte a uvidíte.



## **5.10 Výsledok**

Preloženie udáva tuhosť pružiny zhruba $k \doteq 250~\mathrm{N/m}$. Či toto číslo opisuje dáta, je otázka pre oči, takže preložená priamka ide do grafu. Najprv sa nájdená hodnota dosadí do symbolického modelu:



```sage
# dosadenie číselnej hodnoty parametra k do pôvodného modelu
F_fit(x) = F(x).subs(fit)

# kontrola
F_fit
```

```sage
# graf modelu s nájdenou preloženou hodnotou parametra k
fit_label = rf"Fit, $k = $ {round(fit[k], 2)} N/m"
plot_fit = plot(F_fit, xmin=0, xmax=0.1, legend_label=fit_label, color='orange', linestyle='--') # preloženie bude prerušovaná oranžová čiara
```


> **Poznámka:** Popis je zostavený formátovaním reťazca. Znak `f` pred úvodzovkami umožňuje kódu vkladať hodnoty do textu medzi zloženými zátvorkami, tu preloženú hodnotu `k`, prečítanú zo slovníka ako `fit[k]` a zaokrúhlenú na dve desatinné miesta pomocou `round(fit[k], 2)`. Znak `r` pred úvodzovkami, ako inde, ponechá reťazec surový, takže príkazy $\LaTeX$u v ňom prejdú nedotknuté. Preto je zostavenie popisu zo slovníka lepšie než napísanie čísla do úvodzoviek: keď sa dáta zmenia, nové preloženie zmení `fit[k]` a popis – aj legenda – sa pri ďalšom spustení aktualizujú samy. Ručne napísaný popis povie pravdu raz; formátovaný ju hovorí stále.



```sage
final_plot_fit = plot_fit + err_plot + plot_format
final_plot_fit.show(figsize=12)
```

```sage
# uloženie výsledného grafu
final_plot_fit.save("final_plot_fit.png", figsize=12, dpi=300)
```


To je celá práca so spracovaním dát hotová: sto pochybných údajov dnu, jedno číslo s obrázkom von. Priamka začína tam, kde ju Hookov zákon žiada začať, nulová sila pri nulovom predĺžení, a pretína rozptyl pri každej dĺžke, takže hodnote $k$, ktorú nesie, možno veriť približne na šírku týchto úsečiek.

**Za hranicou priamok.** Priamka prechádzajúca počiatkom je najzhovievavejší model vo fyzike a jej preloženie je rutinná práca: zvládne ju každý tabuľkový procesor, každý nástroj typu „klikni a vyber“. Sila SageMath sa ukazuje inde. Predpokladajme, že fyzika nie je Hookov zákon, ale príťažlivosť medzi dvoma zväzkami magnetov, ktoré sa od seba odťahujú: príťažlivosť je najsilnejšia pri dotyku a so vzdialenosťou slabne a experimentálny vzorec pre ňu znie

$$F(x) = F_0\left(1 - \tanh\frac{x}{a}\right)\, e^{-x/a},$$

hyperbolický tangens tlmený exponenciálou, s dvoma parametrami: mierkou sily $F_0$ a dosahom $a$. Zapísaný v Sage je to len ďalší model:



```sage
# premenné modelu magnetov
F0, a, x = var("F0 a x")

# definovanie modelu
F_magnet(x) = F0 * (1 - tanh(x/a)) * exp(-x/a)

# rýchly pohľad na tvar modelu s vzorovými hodnotami F0 = 3.5 N, a = 1 cm
plot(F_magnet.subs(F0=3.5, a=0.01), xmin=0, xmax=0.05, axes_labels=[r"$x$ [m]", r"$F$ [N]"], frame=True, gridlines=True)
```


Krivka začína vo svojom maxime $F_0$ a hladko klesá k nule tempom, ktoré riadia dva parametre. Taký model nemá položku v rozbaľovacej ponuke nástroja typu „klikni a vyber“, kde bývajú „lineárny“, „polynomický“ a „exponenciálny“ – samotný $\tanh$ je zostavený z exponenciál, $\tanh u = (e^{2u} - 1)/(e^{2u} + 1)$, a preto sa vzorce tohto tvaru často nazývajú *netriviálne*. Presne tu sa symbolický nástroj osvedčí: SageMath preloží takmer každý vzorec, ktorý možno zapísať. Napíšte model, odovzdajte ho funkcii `find_fit` a parametre vypadnú. Rovnaký recept ako pri pružine, len model je ťažší.

> **Poznámka:** Dáta v tejto kapitole boli vygenerované náhodne a slúžia iba na ukážku, ako sa spracúvajú opakované merania tohto druhu. Skutočné pružiny, skutočné diery a skutočné preklepy sa správajú presne rovnako.
