## **6.3 Premenné a funkcie**

| Kód | Čo robí |
|---|---|
| `g = 9.81` | uloží číslo pod meno; jednotka ide do komentára: `# m/s^2` |
| `var('t alpha v0 g')` | deklaruje symbolické premenné: písmená, prázdne miesta |
| `t_h(h) = sqrt(2*h/g)` | uloží vzorec ako symbolickú funkciu premennej `h` |
| `t_hg(h, g) = sqrt(2*h/g)` | funkcia viacerých premenných |
| `t_minutes = t_h/60` | funkcie sú hodnoty: možno ich násobiť, sčítať, kombinovať |
| `show(t_h)` | vysádza vzorec |



```sage
# postup z kapitoly 2: sformulovať model, vyriešiť, dosadiť, čísla nakoniec
var('t alpha v0 g')

x(t) = v0*cos(alpha)*t
y(t) = v0*sin(alpha)*t - 1/2*g*t^2     # model, dve funkcie času

solutions = solve(y(t) == 0, t)        # rovnica vyriešená vzhľadom na t
t_land = solutions[1].rhs()            # výber podľa indexu, odstránenie ľavej strany
show(t_land)

R(alpha) = x(t_land)                   # dosadenie jedného vzorca do druhého
show(R(alpha).trig_reduce())
```


> **Poznámka:** `==` (dve znaky rovnosti) vytvára rovnicu, `=` (jeden) ukladá hodnotu. `solve()` vracia zoznam *rovníc*, nie hodnôt: indexujte pomocou `[i]` a hodnotu získate pomocou `.rhs()`. Funkcia a symbol nemôžu zdieľať písmeno (`x(t)` a `x` kolidujú); náprava je podčiarkovník na konci v kóde, `x_(t) = ...`, ktorý `show()` aj tak vysádza ako obyčajné $x$.



## **6.4 Dosadzovanie**

| Kód | Čo robí |
|---|---|
| `expr.subs(alpha = pi/4)` | dosadí podľa kľúčového slova, jedno alebo viac naraz |
| `values = [y0 == 20, g == 9.81]` | zhromaždí hodnoty parametrov ako zoznam rovníc |
| `expr.subs(values)` | dosadí celý zoznam naraz |
| `values += [m == 500]` | rozšíri zoznam na mieste; možno pridať aj viac rovníc naraz |
| `values_updated = values + [r == 15]` | to isté, ale starý zoznam ostane nedotknutý |

Rovnice sú objekty ako každé iné, takže môžu žiť v zoznamoch a `.subs()` spracuje celý zoznam jedným volaním. Každá rovnica nesie svoju jednotku v komentári, takže `values` slúži zároveň ako tabuľka fyzikálnych vstupov.



## **6.5 Riešenie rovníc a analýza**

| Kód | Čo robí |
|---|---|
| `solve(equation, unknown)` | symbolické riešenie; vracia zoznam rovníc |
| `solve([eq1, eq2, eq3], x, y, z)` | celá sústava rovníc |
| `solutions[1].rhs()` | vyberie druhé riešenie, ponechá jeho pravú stranu |
| `find_root(T(h) == 3.2, 1, 100)` | numerický „lovec“: koreň v danom intervale |
| `diff(expr, x)` | derivácia výrazu `expr` podľa `x` |
| `diff(expr, x, 2)` | druhá derivácia |
| `solve(diff(R(a), a) == 0, a)` | kandidáti na maximum: kde je dotyčnica vodorovná |



```sage
# optimalizácia, recept z kapitoly 2
dR = diff(R(alpha), alpha)             # derivácia ako samostatný vzorec
optimal = solve(dR == 0, alpha)        # vodorovné dotyčnice

show(dR.subs(alpha = pi/6))           # kladná pod kandidátom: ešte rastie
show(dR.subs(alpha = pi/3))           # záporná nad ním: už klesá – maximum
show(diff(R(alpha), alpha, 2).subs(alpha = pi/4))   # druhá derivácia < 0 hovorí to isté
```


> **Poznámka:** Nulová derivácia znamená *vodorovnú dotyčnicu*, nie *maximum*: vrcholy aj doliny ju majú. Druh bodu potvrďte zmenou znamienka derivácie v jeho okolí, znamienkom druhej derivácie alebo vyhodnotením susedných hodnôt (`N(R(40*pi/180))` oproti `N(R(45*pi/180))` oproti `N(R(50*pi/180))`). Použite `find_root`, keď `solve` vzdá, napr. pri zmesi odmocniny a lineárneho člena; chce rovnicu zapísanú pomocou `==`.

## **6.14 Štandardný postup**

1. **Uložte, čo je známe.** `g = 9.81  # m/s^2`
2. **Deklarujte symboly.** `var('t alpha v0 g')`
3. **Sformulujte model.** `y(t) = v0*sin(alpha)*t - 1/2*g*t^2`
4. **Podmienky riešte symbolicky.** `solve(y(t) == 0, t)`, výber pomocou `[i]`, odstránenie ľavej strany pomocou `.rhs()`
5. **Dosaďte vzorce do seba.** `R(alpha) = x(t_land)`
6. **Upravte.** `.trig_reduce()`, `.factor()`, `simplify_full()`
7. **Optimalizujte.** `solve(diff(R, alpha) == 0, alpha)`, potvrďte zmenou znamienka
8. **Čísla nakoniec, v jednom zozname.** `values = [v0 == 15, g == 9.81]`, potom `.subs(values)`
9. **Prejdite na desatinné čísla.** `N(...)`, `round(..., 2)`
